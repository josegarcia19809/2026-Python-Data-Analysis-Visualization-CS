# 🔗 Combinación de DataFrames y Series

### 📚 Usando `pd.concat()` con Series

```python
import pandas as pd
```

```python
s1 = pd.Series(['a', 'b', 'c'])
s2 = pd.Series(['d', 'e', 'f', 'z'])
```

```python
pd.concat([s1,s2])
```

### 🔗 Concatenar dos Series

Esta operación utiliza `pd.concat()` para unir las series `s2` y `s1` en una sola estructura.

Los elementos de `s2` se colocan primero y, a continuación, se agregan los elementos de `s1`, conservando sus índices originales.

El resultado es una nueva serie que contiene todos los valores de ambas series, permitiendo combinar datos que se encuentran separados en distintas estructuras. 📊🔗

```python
pd.concat([s2, s1])
```

### 🔗 Concatenar dos Series y reiniciar los índices

Esta operación utiliza `pd.concat()` para unir las series `s2` y `s1` en una sola serie.

A diferencia de la concatenación estándar, el parámetro `ignore_index=True` hace que Pandas ignore los índices originales de ambas series y genere nuevos índices consecutivos comenzando desde **0**.

El resultado es una serie combinada con una numeración continua, lo que puede facilitar el trabajo con los datos cuando los índices originales ya no son necesarios. 📊🔢

```python
pd.concat([s2, s1], ignore_index=True)
```

```python
nums = pd.Series([65,69,99,23,56,54])
nums2 = pd.Series([11,111,1111])
```

### 🔗 Concatenar múltiples Series y reiniciar los índices

Esta operación utiliza `pd.concat()` para unir las series `s2`, `s1` y `nums` en una sola serie.

El parámetro `ignore_index=True` indica que los índices originales deben descartarse y reemplazarse por una nueva secuencia de índices consecutivos que comienza en **0**.

El resultado es una única serie que contiene todos los valores de las tres series en el orden en que fueron proporcionadas, facilitando el manejo de los datos como una sola colección continua. 📊🔢

```python
pd.concat([s2, s1, nums], ignore_index=True)
```

```python
pd.concat([nums, nums2, s1], ignore_index=True)
```

### 📚 Usando `pd.concat()` con Series by Index

```python
c1 = pd.Series(["red", "orange", "yellow"])
c2 = pd.Series(["green", "blue", "purple"])
```

```python
pd.concat([c1,c2])
```

### 📊 Concatenar Series por columnas

Esta operación utiliza `pd.concat()` para combinar las series `c1` y `c2` horizontalmente mediante `axis=1`.

En lugar de agregar los valores uno debajo de otro, las series se colocan lado a lado como columnas de un nuevo DataFrame.

Pandas alinea los datos utilizando los índices de ambas series. Si un índice existe en una serie pero no en la otra, se mostrarán valores faltantes (`NaN`) en las posiciones correspondientes.

El resultado es un DataFrame que reúne la información de ambas series en una estructura tabular. 📋🔗

```python
pd.concat([c1,c2], axis=1)
```

### 🍎🐾 Crear dos Series con índices personalizados

Se crean dos objetos `Series` llamados `fruits` y `animals`, cada uno con tres elementos y con índices personalizados.

La serie **`fruits`** asocia cada índice con el nombre de una fruta, mientras que la serie **`animals`** asocia los mismos índices con nombres de animales.

Aunque ambas series contienen los índices `a`, `b` y `c`, estos aparecen en diferente orden. Esta característica es importante porque Pandas utiliza los índices para alinear los datos cuando se realizan operaciones entre series, como sumas, concatenaciones o combinaciones.

Este ejemplo ayuda a comprender cómo Pandas relaciona los datos mediante sus etiquetas de índice en lugar de depender únicamente de la posición de los elementos. 📊🔗

```python
fruits = pd.Series(
    data=["apple", "banana", "cherry"], 
    index=["a","b", "c"]
)

animals = pd.Series(
    data=["badger", "cougar", "anaconda"], 
    index=["b", "c", "a"]
)
```

```python
fruits
```

```python
animals
```

### 📊 Combinar Series por columnas utilizando sus índices

Esta operación utiliza `pd.concat()` con `axis=1` para colocar las series `fruits` y `animals` una junto a la otra, formando un nuevo DataFrame.

Pandas alinea automáticamente los datos según los índices de cada serie. Aunque los elementos aparecen en distinto orden dentro de las series originales, los valores se emparejan correctamente porque comparten los mismos índices (`a`, `b` y `c`).

El resultado es un DataFrame donde cada fila reúne la fruta y el animal asociados al mismo índice, demostrando cómo Pandas utiliza las etiquetas de índice para organizar y combinar datos de manera inteligente. 🍎🐾📊

```python
pd.concat([fruits, animals], axis=1)
```

### 📊 Combinar Series por columnas asignando nombres a las columnas

Esta operación utiliza `pd.concat()` para unir las series `fruits` y `animals` horizontalmente (`axis=1`) y crear un DataFrame.

El parámetro `keys` permite asignar nombres a las columnas resultantes. En este caso, las columnas se etiquetan como **`fruit`** y **`animal`**, facilitando la identificación de la información que contiene cada una.

Pandas alinea automáticamente los datos utilizando los índices de las series, asegurando que los valores correspondientes queden agrupados en la misma fila.

El resultado es un DataFrame más descriptivo y fácil de interpretar, ya que cada columna cuenta con una etiqueta significativa. 🍎🐾📋

```python
pd.concat([fruits, animals], axis=1, keys=["fruit", "animal"])
```

### 🔗 Concatenar Series por filas utilizando claves

Esta operación utiliza `pd.concat()` para unir las series `fruits` y `animals` verticalmente (`axis=0`), es decir, una debajo de la otra.

El parámetro `keys` agrega un nivel adicional al índice, indicando de qué serie proviene cada conjunto de datos. En este caso, los elementos quedarán agrupados bajo las etiquetas **`fruit`** y **`animal`**.

El resultado es una única serie con un índice jerárquico (*MultiIndex*), que permite identificar tanto la serie de origen como el índice original de cada elemento.

Esta técnica es útil cuando se desea combinar varios conjuntos de datos sin perder información sobre su procedencia. 🍎🐾📊

```python
pd.concat([fruits, animals], axis=0, keys=["fruit", "animal"])
```

### 🧱 Inner vs Outer Joins

```python
fruits = pd.Series(
    data=["apple", "banana", "cherry", "durian"],
    index=["a","b", "c", "d"]
)

animals = pd.Series(
    data=["badger", "cougar", "anaconda", "elk", "pika"],
    index=["b", "c", "a", "e", "p"]
)
```

```python
pd.concat([animals, fruits], ignore_index=True)
```

```python
pd.concat([animals, fruits], axis=1)
```

### 📊 Combinar Series por columnas utilizando una unión interna

Esta operación utiliza `pd.concat()` para unir las series `animals` y `fruits` horizontalmente (`axis=1`).

El parámetro `join="inner"` indica que solo se conservarán los índices que estén presentes en ambas series. De esta forma, el resultado incluirá únicamente las filas cuyos índices sean comunes a todos los objetos concatenados.

Como ambas series comparten los índices `a`, `b` y `c`, estos serán los únicos que aparecerán en el DataFrame resultante.

Esta opción es útil cuando se desea trabajar únicamente con los datos que existen en todas las estructuras involucradas en la combinación. 📊🔗

```python
pd.concat([animals, fruits], axis=1, join="inner")
```

## Concatenating DataFrames By Columns

### 🥔🌱 Crear dos DataFrames con información de cosechas

Se crean dos DataFrames llamados `harvest_21` y `harvest_22`, cada uno con datos de productos agrícolas y sus cantidades cosechadas.

Ambos DataFrames contienen dos columnas:

- 🌾 `crop`: nombre del cultivo.
- 📦 `qty`: cantidad cosechada.

El DataFrame `harvest_21` almacena información correspondiente a una cosecha, mientras que `harvest_22` contiene datos de otra cosecha diferente.

Algunos cultivos aparecen en ambos conjuntos de datos, como **garlic** y **onions**, mientras que otros son exclusivos de una sola cosecha. Esto permite realizar posteriormente operaciones de concatenación, combinación o comparación para analizar la evolución de las cantidades cosechadas entre ambos periodos. 📊🚜

```python
harvest_21 = pd.DataFrame(
    [['potatoes', 900], ['garlic', 1350], ['onions', 875]], 
    columns=['crop', 'qty']
)

harvest_22 = pd.DataFrame(
    [['garlic', 1600], ['spinach', 560], ['turnips', 999], ['onions', 1000]], 
    columns=['crop', 'qty']
)
```

```python
harvest_21
```

```python
harvest_22
```

### 📊 Unir dos DataFrames y reiniciar los índices

Esta operación utiliza `pd.concat()` para combinar los DataFrames `harvest_21` y `harvest_22` verticalmente, agregando las filas de un DataFrame debajo de las del otro.

El parámetro `ignore_index=True` indica que los índices originales deben descartarse y reemplazarse por una nueva secuencia de índices consecutivos que comienza en **0**.

El resultado es un único DataFrame que contiene todos los registros de ambas cosechas, facilitando el análisis conjunto de la información sin preocuparse por posibles índices duplicados o desordenados. 🌾📦📊

```python
pd.concat([harvest_21, harvest_22], ignore_index=True)
```

### 📊 Unir dos DataFrames utilizando claves

Esta operación utiliza `pd.concat()` para combinar los DataFrames `harvest_21` y `harvest_22` verticalmente.

El parámetro `keys` agrega un nivel adicional al índice, permitiendo identificar el origen de cada fila. En este caso, las claves **2021** y **2022** indican a qué cosecha pertenece cada registro.

El resultado es un DataFrame con un índice jerárquico (*MultiIndex*), donde el primer nivel representa el año de la cosecha y el segundo nivel corresponde al índice original de cada fila.

Esta técnica es útil para conservar información sobre la procedencia de los datos después de combinarlos en una sola estructura. 🌾📦🗓️

```python
pd.concat([harvest_21, harvest_22], keys=[2021, 2022])
```

### 🌾📦💰 Crear un DataFrame con cantidades y ganancias de cosecha

Se crea un DataFrame llamado `harvest_23` que almacena información sobre distintos cultivos, la cantidad cosechada y la ganancia obtenida.

El DataFrame contiene tres columnas:

- 🌱 `crop`: nombre del cultivo.
- 📦 `qty`: cantidad cosechada.
- 💰 `profit`: ganancia asociada a ese cultivo.

Cada fila representa un cultivo diferente junto con sus datos de producción y rentabilidad.

Este tipo de estructura permite analizar no solo la cantidad producida, sino también el beneficio económico generado por cada cultivo, facilitando comparaciones y estudios de rendimiento agrícola. 📊🚜💵

```python
harvest_23 = pd.DataFrame(
    [['potatoes', 900, 500], ['garlic', 1350, 1200], ['onions', 875, 950]], 
    columns=['crop', 'qty', 'profit']
)
```

### 📊 Unir varios DataFrames con diferentes columnas

Esta operación utiliza `pd.concat()` para combinar verticalmente los DataFrames `harvest_21`, `harvest_22` y `harvest_23`.

Pandas incluye todas las columnas que aparecen en cualquiera de los DataFrames. Cuando una fila proviene de un DataFrame que no contiene alguna de esas columnas, se asigna un valor faltante (`NaN`) en la posición correspondiente.

En este caso, la columna **`profit`** solo existe en `harvest_23`, por lo que las filas provenientes de `harvest_21` y `harvest_22` tendrán valores faltantes en dicha columna.

El resultado es un único DataFrame que reúne toda la información disponible, conservando los datos de producción y, cuando existan, las ganancias asociadas a cada cultivo. 🌾📦💰

```python
pd.concat([harvest_21, harvest_22, harvest_23])
```

### 📊 Unir varios DataFrames conservando solo las columnas comunes

Esta operación utiliza `pd.concat()` para combinar verticalmente los DataFrames `harvest_21`, `harvest_22` y `harvest_23`.

El parámetro `join="inner"` indica que únicamente deben conservarse las columnas que están presentes en **todos** los DataFrames involucrados.

Como la columna **`profit`** solo existe en `harvest_23`, esta será excluida del resultado. Únicamente se mantendrán las columnas comunes, como **`crop`** y **`qty`**.

El resultado es un DataFrame unificado que contiene únicamente la información compartida por todos los conjuntos de datos, evitando la aparición de valores faltantes (`NaN`) debido a diferencias en la estructura de las columnas. 🌾📦📊

```python
pd.concat([harvest_21, harvest_22, harvest_23], join="inner")
```

## 📌 Concatenación de DataFrames por Índice

### 🐾📊 Crear dos DataFrames relacionados con animales

Se crean dos DataFrames llamados `livestock` y `weights`, ambos utilizando nombres de animales como índices.

#### 🏠 `livestock`
Contiene información sobre los animales y el lugar donde se encuentran, además de la cantidad registrada de cada uno.

Sus columnas son:

- 📍 `location`: ubicación del animal.
- 🔢 `qty`: cantidad registrada.

#### ⚖️ `weights`
Contiene información sobre los pesos mínimo y máximo asociados a cada tipo de animal.

Sus columnas son:

- ⚖️ `min_weight`: peso mínimo.
- 🏋️ `max_weight`: peso máximo.

Los índices de ambos DataFrames representan especies de animales. Algunos animales aparecen en ambos conjuntos de datos, como **alpaca**, **horse** y **chicken**, mientras que otros solo están presentes en uno de ellos.

Esta estructura es ideal para realizar operaciones de combinación (*merge* o *join*), ya que los índices pueden utilizarse para relacionar la información de ubicación, cantidad y peso de cada animal. 🐴🐔🦙📊

```python
livestock = pd.DataFrame(
    [['pasture', 9], ['stable', 3], ['coop', 34]], 
    columns=['location', 'qty'],
    index=['alpaca', 'horse', 'chicken']
)
weights = pd.DataFrame(
    [[4,10], [900, 2000], [1.2, 4], [110, 150]], 
    columns=['min_weight', 'max_weight'],
    index=['chicken', 'horse', 'duck', 'alpaca']
)
```

```python
livestock
```

```python
weights
```

### 📊 Unir DataFrames con columnas diferentes

Esta operación utiliza `pd.concat()` para combinar verticalmente los DataFrames `livestock` y `weights`, agregando las filas de uno debajo de las del otro.

Como ambos DataFrames tienen columnas distintas, Pandas crea un conjunto de columnas que incluye todas las existentes en ambos. Cuando una fila proviene de un DataFrame que no contiene alguna de esas columnas, se asigna un valor faltante (`NaN`).

El resultado es un único DataFrame que reúne información de ubicación, cantidad y pesos de los animales, aunque algunas celdas contendrán valores faltantes debido a las diferencias en la estructura original de los datos.

Esta forma de concatenación es útil para combinar conjuntos de datos relacionados, incluso cuando no comparten exactamente las mismas columnas. 🐔🐴🦙📊

```python
pd.concat([livestock, weights])
```

### 📊 Combinar DataFrames por columnas utilizando los índices

Esta operación utiliza `pd.concat()` con `axis=1` para unir los DataFrames `livestock` y `weights` horizontalmente, colocando sus columnas lado a lado.

Pandas alinea automáticamente los datos utilizando los índices de ambos DataFrames. Los animales que aparecen en ambos conjuntos de datos tendrán toda su información en la misma fila.

Si un animal existe en un DataFrame pero no en el otro, las columnas faltantes se completarán con valores `NaN`.

El resultado es un nuevo DataFrame que reúne información sobre ubicación, cantidad y rango de peso para cada animal, aprovechando los índices como criterio de correspondencia. 🐔🐴🦙⚖️📊

```python
pd.concat([livestock, weights], axis=1)
```

### 📊 Combinar DataFrames por columnas conservando solo los índices comunes

Esta operación utiliza `pd.concat()` para unir los DataFrames `livestock` y `weights` horizontalmente (`axis=1`).

El parámetro `join="inner"` indica que únicamente se conservarán las filas cuyos índices estén presentes en ambos DataFrames.

En este caso, solo aparecerán los animales que existen tanto en `livestock` como en `weights`, mientras que aquellos que estén presentes en uno solo de los DataFrames serán excluidos.

El resultado es un DataFrame que combina la información de ubicación, cantidad y peso únicamente para los animales compartidos por ambos conjuntos de datos, evitando la generación de valores faltantes (`NaN`). 🐔🐴🦙📊

```python
pd.concat([livestock, weights], axis=1, join="inner")
```
