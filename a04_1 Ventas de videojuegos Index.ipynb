# 🧠 Pandas Indexes

Los **índices (Indexes)** son una de las características más importantes de **Pandas**. Cada fila de un `DataFrame` o una `Series` tiene un índice que funciona como una **etiqueta única**, permitiendo identificar, localizar y manipular los datos de forma rápida y eficiente.

Aunque Pandas crea automáticamente un índice numérico al cargar un conjunto de datos, también es posible utilizar una o varias columnas como índice. Esto facilita las búsquedas, el filtrado y el acceso a registros específicos, especialmente cuando se trabaja con grandes volúmenes de información.

## 🎯 ¿Qué aprenderás?

En esta sección aprenderás a:

- 📌 Comprender qué es un **índice** y cuál es su función.
- 📌 Visualizar los índices de un `DataFrame` y de una `Series`.
- 📌 Establecer una columna como índice utilizando `set_index()`.
- 📌 Restablecer el índice con `reset_index()`.
- 📌 Acceder a registros mediante sus etiquetas de índice.
- 📌 Entender las ventajas de utilizar índices para organizar los datos.

## 💡 Ejemplo de uso

Imagina un conjunto de datos con información de estudiantes:

| Matrícula | Nombre | Carrera | Promedio |
|-----------|---------|----------|-----------|
| A001 | Ana | Informática | 9.4 |
| A002 | Luis | Civil | 8.7 |
| A003 | Sofía | Química | 9.8 |

En lugar de utilizar los índices numéricos (`0`, `1`, `2`), puedes establecer la columna **Matrícula** como índice. De esta manera, será más sencillo localizar un estudiante utilizando su matrícula, en lugar de recordar el número de la fila.

> 🚀 Dominar el uso de los **Indexes** te permitirá trabajar con datos de forma más organizada, realizar búsquedas más eficientes y aprovechar al máximo las herramientas de análisis que ofrece Pandas.

## 🔎 Fundamentos de Index

En esta sección vamos a:
- Cargar un archivo CSV con datos de ventas de videojuegos. 🎮
- Establecer un índice personalizado (columna Título). 🔖
- Visualizar datos específicos y graficarlos. 📈

```python
# 📥 Paso 1: Importamos la librería pandas
import pandas as pd
```

```python
# 📂 Paso 2: Cargamos los datos desde el archivo CSV
ventas = pd.read_csv("data/ventas_videojuegos.csv")
```

```python
# 🏷️ Paso 3: Establecemos la columna 'Titulo' como índice
# Esto nos permite buscar directamente por título sin tener que usar índices numéricos
ventas.set_index("Titulo", inplace=True)
ventas
```

```python
# 🔍 Paso 4: Mostramos los índices actuales (los títulos de los videojuegos)
ventas.index
```

```python
# 💵 Paso 5: Mostramos las ventas en Norteamérica para cada videojuego
ventas["Ventas_NA"]
```

```python
# 📈 Paso 6: Graficamos las ventas en Norteamérica para tener una visualización general
ventas["Ventas_NA"].plot()
```

```python
# 🔝 Paso 7: Graficamos solo los primeros 4 videojuegos para un análisis más detallado
ventas["Ventas_NA"].head(4).plot()
```

```python
# 🔄 EXTRA: También podemos establecer el índice desde que cargamos el archivo con index_col
# Esto ahorra el paso de set_index después de cargar el archivo
ventas = pd.read_csv("data/ventas_videojuegos.csv", index_col=0)
ventas
```

```python
# 🧬 Paso 8: Visualizamos la columna 'Genero' para conocer la categoría de cada videojuego
ventas["Genero"]
```
