# 🔽 Ordenamiento en Pandas

En el análisis de datos es muy común necesitar **ordenar la información** para identificar los valores más altos, los más bajos o simplemente presentar los datos de una manera más clara. **Pandas** ofrece herramientas muy sencillas para ordenar filas y columnas según uno o varios criterios.

El ordenamiento permite explorar mejor los datos, detectar tendencias, encontrar registros destacados y preparar la información para generar reportes o visualizaciones.

## 🎯 ¿Qué aprenderás?

En esta sección aprenderás a:

- 📌 Ordenar un `DataFrame` por una o varias columnas.
- 📌 Ordenar datos de forma ascendente y descendente.
- 📌 Ordenar utilizando el índice del `DataFrame`.
- 📌 Conservar o modificar el `DataFrame` original al ordenar.
- 📌 Aplicar el ordenamiento para facilitar el análisis de datos.

## 💡 Ejemplo de uso

Imagina un conjunto de datos con el promedio de varios estudiantes:

| Nombre | Carrera | Promedio |
|---------|----------|-----------|
| Ana | Informática | 9.4 |
| Luis | Civil | 8.7 |
| Sofía | Química | 9.8 |
| Carlos | Industrial | 8.9 |

Si ordenas la columna **Promedio** de forma descendente, podrás identificar rápidamente a los estudiantes con las calificaciones más altas. También podrías ordenar por **Carrera** para agrupar a los estudiantes de la misma especialidad.

> 🚀 El ordenamiento es una de las operaciones más utilizadas en Pandas, ya que facilita la exploración, comparación y presentación de los datos, permitiendo obtener información útil de manera rápida y eficiente.

📥 Importamos la librería `pandas` para trabajar con estructuras de datos como DataFrames.

```python
import pandas as pd
```

📂 Leemos el archivo CSV que contiene los datos de ventas de videojuegos y lo almacenamos en un DataFrame llamado `ventas`.

```python
ventas = pd.read_csv("data/ventas_videojuegos.csv")
```

📊 Ordenamos el DataFrame por la columna `Ventas_Globales` en orden ascendente (de menor a mayor).

```python
ventas.sort_values("Ventas_Globales")
```

📈 Ordenamos el DataFrame por la columna `Ventas_Globales` en orden descendente (de mayor a menor).

```python
ventas.sort_values("Ventas_Globales", ascending=False)
```

👁️ Mostramos el DataFrame original para ver el estado actual de los datos.

```python
ventas
```

📝 Ordenamos el DataFrame por `Ventas_Globales` en orden descendente y aplicamos el cambio directamente usando `inplace=True`.

```python
ventas.sort_values("Ventas_Globales", ascending=False, inplace=True)
ventas
```

🧮 Ordenamos el DataFrame por dos columnas: `Año` y `Genero`, en orden ascendente.

```python
ventas.sort_values(["Año", "Genero"], inplace=True)
ventas
```

🔁 Ordenamos por `Año` y `Genero`, pero en orden descendente esta vez.

```python
ventas.sort_values(["Año", "Genero"], ascending=False, inplace=True)
ventas
```

🏷️ Establecemos la columna `Titulo` como índice del DataFrame para facilitar búsquedas por título.

```python
ventas.set_index("Titulo", inplace=True)
ventas
```

📌 Ordenamos el índice (los títulos de los videojuegos) en orden alfabético ascendente.

```python
ventas.sort_index()
```

🔽 Ahora ordenamos el índice (los títulos) en orden descendente directamente en el DataFrame.

```python
ventas.sort_index(ascending=False, inplace=True)
ventas
```

🔢 Mostramos la frecuencia de cada valor en la columna `Ventas_Globales` usando `value_counts()`.

```python
ventas["Ventas_Globales"].value_counts()
```

📊 Graficamos un histograma de las frecuencias de ventas globales con barras verticales (`bar`).

```python
ventas["Ventas_Globales"].value_counts().plot(kind="bar")
```

🔽 Ordenamos las frecuencias de ventas de menor a mayor antes de graficar.

```python
ventas["Ventas_Globales"].value_counts().sort_values()
```

📉 Graficamos las ventas ordenadas de menor a mayor para observar mejor la distribución.

```python
ventas["Ventas_Globales"].value_counts().sort_values().plot(kind="bar")
```

🔢 Ahora ordenamos los valores únicos de `Ventas_Globales` por el índice (es decir, por el valor numérico en sí).

```python
ventas["Ventas_Globales"].value_counts().sort_index()
```

📈 Graficamos la frecuencia de ventas, esta vez ordenadas por el índice (valor de ventas).

```python
ventas["Ventas_Globales"].value_counts().sort_index().plot(kind="bar")
```

🎮 Finalmente, graficamos la frecuencia de cada género de videojuegos, ordenado por orden alfabético del género.

```python
ventas["Genero"].value_counts().sort_index().plot(kind="bar")
```
