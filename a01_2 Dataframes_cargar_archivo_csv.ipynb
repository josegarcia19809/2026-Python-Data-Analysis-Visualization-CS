# 📊 Aspectos básicos de los DataFrames con Pandas

Este notebook muestra cómo trabajar con DataFrames utilizando la biblioteca **Pandas** en Python. Aprenderás a:

- Cargar un archivo CSV.
- Visualizar registros.
- Consultar información básica del DataFrame.
- Conocer su estructura (filas, columnas, tipos de datos).

```python
import pandas as pd
```

## 📥 Cargando datos desde un archivo CSV

Utilizamos `pd.read_csv()` para leer un archivo CSV y convertirlo automáticamente en un DataFrame, que es una estructura de datos tabular (como una hoja de cálculo).

```python
estados = pd.read_csv("data/estados.csv")
```

## 👀 Explorando el contenido del DataFrame

Podemos visualizar el contenido completo del DataFrame (si no es muy grande) simplemente escribiendo su nombre.

```python
# Mostrar el contenido del dataframe
estados
```

## 🔍 Vistazos rápidos con `head()` y `tail()`

- `head(n)` muestra las primeras `n` filas del DataFrame.
- `tail(n)` muestra las últimas `n` filas.

Estas funciones son útiles para explorar cómo están estructurados los datos al inicio y al final.


### head() muestra los primeros 5 renglones

```python
estados.head()
```

### head(12) muestra los primeros 12 renglones

```python
estados.head(12)
```

### tail() muestra los últimos 5 renglones

```python
estados.tail()
```

### tail(9) muestra los últimos 9 renglones

```python
estados.tail(9)
```

## 🏷️ Nombres de las columnas

Los nombres de las columnas se almacenan como un objeto tipo `Index`. Esto permite ver fácilmente qué atributos contiene cada registro.


```python
estados.columns
```

### Mostrando información del Dataframe

## 🧠 Inspección general del DataFrame con `info()`

El método `estados.info()` nos da un resumen muy útil del DataFrame. Muestra:

- El número total de entradas (filas).
- El nombre y tipo de cada columna.
- Cuántos valores **no nulos** hay por columna.
- La **cantidad total de memoria** usada.

Esto es útil para:
- Detectar columnas con valores faltantes (null).
- Entender qué tipo de datos contiene cada columna.
- Comprobar si es necesario transformar datos (por ejemplo, cambiar de `object` a `int` o `datetime`).


```python
estados.info()
```

## 🧾 Tipos de datos de cada columna con `dtypes`

El atributo `estados.dtypes` nos muestra qué tipo de dato tiene cada columna del DataFrame. Los tipos más comunes en Pandas son:

- `object`: Texto (cadenas de caracteres).
- `int64`: Números enteros.
- `float64`: Números decimales.
- `bool`: Valores lógicos (True/False).
- `datetime64`: Fechas y horas.

🔎 **¿Para qué sirve esto?**
Conocer los tipos de datos es útil para:

- Saber si necesitas convertir algún dato (por ejemplo, texto a fecha).
- Aplicar funciones adecuadas según el tipo (por ejemplo, operaciones matemáticas en `int` o `float`).
- Identificar posibles errores en la carga de datos (como fechas que se cargaron como `object`).


### Mostrando los tipos de datos que guarda cada columna

```python
estados.dtypes
```

## 🧾 Tamaño y estructura del DataFrame

- `type(estados)`: Muestra el tipo de objeto (debe ser `<class 'pandas.core.frame.DataFrame'>`).
- `len(estados)`: Cantidad de filas (registros).
- `estados.shape`: Una tupla con el número de filas y columnas.
- `estados.size`: Número total de elementos (filas × columnas).


```python
type(estados)
```

### Mostrando la cantidad de registros

```python
len(estados)
```

### Mostrando la cantidad de registros y columnas

```python
estados.shape
```

### Mostrando la cantidad de casillas de todo el DataFrame

```python
estados.size
```
