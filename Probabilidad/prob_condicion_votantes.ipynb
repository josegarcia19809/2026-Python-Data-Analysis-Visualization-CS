```python
import pandas as pd
```

```python
votantes = pd.read_csv("../data/votantes_13660_edades.csv")
votantes.head()
```

```python
# Definir los límites de los rangos
limites = [17, 29, 44, 64, float("inf")]

# Etiquetas de los rangos
etiquetas = [
    "18 a 29",
    "30 a 44",
    "45 a 64",
    "65 y más"
]

# Crear la nueva columna
votantes["rango_edad"] = pd.cut(
    votantes["edad"],
    bins=limites,
    labels=etiquetas
)
```

```python
votantes.head()
```

```python
tabla_edades = votantes["rango_edad"].value_counts().sort_index()
print(tabla_edades)
```

```python
tabla_edad_voto = pd.crosstab(
    votantes["rango_edad"],
    votantes["voto"],
    margins=True,
    margins_name="Total"
)
print(tabla_edad_voto)
```

```python
prob_18_29 = (tabla_edad_voto.loc["18 a 29", "Total"] / tabla_edad_voto.loc["Total", "Total"])
prob_30_44 = (tabla_edad_voto.loc["30 a 44", "Total"] / tabla_edad_voto.loc["Total", "Total"])
prob_45_64 = (tabla_edad_voto.loc["45 a 64", "Total"] / tabla_edad_voto.loc["Total", "Total"])
prob_65_mas = (tabla_edad_voto.loc["65 y más", "Total"] / tabla_edad_voto.loc["Total", "Total"])

print("18 a 29:", (prob_18_29 * 100).round(0), "%")
print("30 a 44:", (prob_30_44 * 100).round(0), "%")
print("45 a 64:", (prob_45_64 * 100).round(0), "%")
print("65 y más:", (prob_65_mas * 100).round(0), "%")
```

```python
tabla_genero_voto = pd.crosstab(
    votantes["genero"],
    votantes["voto"],
    margins=True,
    margins_name="Total"
)
print(tabla_genero_voto)
```

```python
prob_mujer_mcCain = (tabla_genero_voto.loc["Mujer", "McCain"] / tabla_genero_voto.loc["Mujer", "Total"])
print((prob_mujer_mcCain * 100).round(0))
```

```python
prob_hombre_mcCain = (tabla_genero_voto.loc["Hombre", "McCain"] / tabla_genero_voto.loc["Hombre", "Total"])
print((prob_hombre_mcCain * 100).round(0))
```

```python
prob_hombre = (tabla_genero_voto.loc["Hombre"] / tabla_genero_voto.loc["Total"])
print((prob_hombre * 100).round(0))
```

```python
prob_hombre = (
    tabla_genero_voto.loc["Hombre", "Total"] /
    tabla_genero_voto.loc["Total", "Total"]
)

prob_mujer = (
    tabla_genero_voto.loc["Mujer", "Total"] /
    tabla_genero_voto.loc["Total", "Total"]
)



print("Hombre:", (prob_hombre * 100).round(0), "%")
print("Mujer:", (prob_mujer * 100).round(0), "%")

print("Suma:", prob_hombre + prob_mujer)
```
