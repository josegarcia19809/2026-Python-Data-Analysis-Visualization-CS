```python
import pandas as pd
```

```python
alumnos = pd.read_csv("../data/estudiantes_478_registros.csv")
alumnos
```

```python
agrupacion = alumnos.groupby(["Sexo", "Objetivo_principal"]).size()
agrupacion
```

```python
tabla = pd.crosstab(
    alumnos["Sexo"], alumnos["Objetivo_principal"],
    margins=True,
    margins_name= "Total"
)
tabla
```

### 4. 📊 Calcular probabilidades condicionales

Por ejemplo, ¿cuál es la probabilidad de que un estudiante quiera destacar en deportes, sabiendo que es niña?

**P(Deportes | Niña) = 30 / 251 ≈ 0.1195 = 11.95 %**

```python
probabilidad = (
    tabla.loc["Niñas", "Deportes"]
    / tabla.loc["Niñas", "Total"]
)

print(f"Probabilidad: {probabilidad:.3f}")
print(f"Porcentaje: {probabilidad:.1%}")
```
