```python
import pandas as pd
```

```python
votantes = pd.read_csv("../data/votantes_13660_edades.csv")
votantes.head()
```

```python
votantes_18_29_mc_cain = votantes[(votantes["edad"] >= 18) & (votantes["edad"] <= 29) & (votantes["voto"] == "McCain")]
print(votantes_18_29_mc_cain.shape[0])
```

```python
# Total de votantes entre 18 y 29 años
total_18_29 = votantes[
    (votantes["edad"] >= 18) &
    (votantes["edad"] <= 29)
    ].shape[0]

# Votantes entre 18 y 29 años que votaron por McCain
total_mcCain_18_29 = votantes_18_29_mc_cain.shape[0]

# Calcular la probabilidad condicional
probabilidad = total_mcCain_18_29 / total_18_29

print(f"Probabilidad: {probabilidad * 100:.1f}%")
```

```python
votantes_18_29 = votantes[(votantes["edad"] >= 18) & (votantes["edad"] <= 29)]
print(votantes_18_29.shape[0])
print(votantes_18_29.shape[0] / votantes.shape[0] * 100)
```

```python
votantes_30_44 = votantes[(votantes["edad"] >= 30) & (votantes["edad"] <= 44)]
print(votantes_30_44.shape[0])
print(votantes_30_44.shape[0] / votantes.shape[0] * 100)
```

```python
votantes_45_64 = votantes[(votantes["edad"] >= 45) & (votantes["edad"] <= 64)]
print(votantes_45_64.shape[0])
print(votantes_45_64.shape[0] / votantes.shape[0] * 100)
```

```python
votantes_65_mas = votantes[(votantes["edad"] >= 65)]
print(votantes_65_mas.shape[0])
print(votantes_65_mas.shape[0] / votantes.shape[0] * 100)
```

```python
print("18 a 29 McCain: ", votantes_18_29[votantes_18_29["voto"] == "McCain"].shape[0])
print("30 a 44 McCain: ", votantes_30_44[votantes_30_44["voto"] == "McCain"].shape[0])
print("45 a 64 McCain: ", votantes_45_64[votantes_45_64["voto"] == "McCain"].shape[0])
print("65 o más McCain: ", votantes_65_mas[votantes_65_mas["voto"] == "McCain"].shape[0])
```

```python
print("18 a 29 Obama: ", votantes_18_29[votantes_18_29["voto"] == "Obama"].shape[0])
print("30 a 44 Obama: ", votantes_30_44[votantes_30_44["voto"] == "Obama"].shape[0])
print("45 a 64 Obama: ", votantes_45_64[votantes_45_64["voto"] == "Obama"].shape[0])
print("65 o más Obama: ", votantes_65_mas[votantes_65_mas["voto"] == "Obama"].shape[0])
```
