# 🔍 Filtrado de datos usando Series Booleanas
Aprenderemos a usar condiciones simples para filtrar registros en un DataFrame utilizando pandas 🐼.

### 📥 1. Importar la librería pandas
Usamos pandas para manejar datos en forma de tablas (DataFrames).

```python
import pandas as pd
```

### 📄 2. Cargar el archivo CSV del Titanic
Leemos el archivo `titanic_v5.csv` con los datos de los pasajeros 🛳️.

```python
titanic = pd.read_csv("data/titanic_v5.csv")
```

### 👀 3. Visualizar el DataFrame completo
Veamos una vista general del conjunto de datos cargado.

```python
titanic
```

### 🙋‍♀️ 4. Visualizar la columna `sexo`
Mostramos solo los valores de la columna `sexo` para entender su contenido.

```python
titanic["sexo"]
```

### ✅ 5. Crear una serie booleana: mujeres
Esto devuelve `True` o `False` dependiendo si el valor es `"female"`.

```python
titanic["sexo"] == "female"
```

### 👩 6. Filtrar mujeres pasajeras
Usamos la condición booleana anterior para mostrar solo a las mujeres.

```python
# Mostrar a las personas de sexo femenino ("female")
titanic[titanic["sexo"] == "female"]
```

### ❌ 7. Filtrar pasajeros que no sobrevivieron
Filtramos las filas donde la columna `sobrevivio` es igual a 0.

```python
# Mostrar los pasajeros que no sobrevivieron
titanic[titanic["sobrevivio"] == 0]
```

### ✅ 8. Filtrar pasajeros que sobrevivieron
Ahora mostramos las filas donde `sobrevivio` es igual a 1.

```python
# Mostrar los pasajeros que sobrevivieron
titanic[titanic["sobrevivio"] == 1]
```

### 👶 9. Filtrar por edad específica (18 años)
Mostramos los registros donde la edad es exactamente 18.

```python
# Mostrar a los pasajeros cuya edad sea 18
titanic[titanic["edad"] == 18]
```

### ✈️ 10. Filtrar pasajeros que **no** viajaban en primera clase
Se muestran los pasajeros cuya clase no es igual a 1 (primera clase).

```python
# Mostrar a los pasajeros que no iban en primera clase
titanic[titanic["clase_pasajero"] != 1]
```
