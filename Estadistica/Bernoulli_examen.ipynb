```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Crear el dataset
datos = {
    "Estudiante": [
        f"Estudiante_{i:02d}"
        for i in range(1, 31)
    ],
    "Calificacion": [
        85, 62, 74, 91, 58, 69, 77, 83, 45, 70,
        96, 54, 88, 66, 72, 79, 61, 93, 68, 81,
        57, 75, 84, 49, 71, 64, 90, 73, 59, 87
    ]
}

df = pd.DataFrame(datos)
df
```

```python
# 2. Crear la variable aleatoria de Bernoulli
df["X"] = (df["Calificacion"] >= 70).astype(int)
df["X"]
```

```python
# 3. Agregar una descripción del resultado
df["Resultado"] = df["X"].map({
    1: "Aprobó",
    0: "No aprobó"
})

# 4. Mostrar los datos
print(df.to_string(index=False))
```

```python
# 5. Analizar la variable aleatoria
frecuencias = df["X"].value_counts().sort_index()
proporciones = df["X"].value_counts(normalize=True).sort_index()

print("\nFrecuencias:")
print(frecuencias.rename(index={
    0: "No aprobó",
    1: "Aprobó"
}))

print("\nProporciones:")
print(proporciones.rename(index={
    0: "No aprobó",
    1: "Aprobó"
}))
```

```python
# 6. Calcular la probabilidad observada de aprobar
p = df["X"].mean()

print(f"\nProporción de aprobados: {p:.2%}")
print(f"Calificación promedio: {df['Calificacion'].mean():.2f}")
```

```python
# 7. Graficar los resultados
df["Resultado"].value_counts().reindex(
    ["No aprobó", "Aprobó"], fill_value=0
).plot(kind="bar")

plt.title("Resultados del examen")
plt.xlabel("Resultado")
plt.ylabel("Número de estudiantes")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
