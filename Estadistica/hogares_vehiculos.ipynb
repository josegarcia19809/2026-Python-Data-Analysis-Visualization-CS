```python
import pandas as pd
```

```python
vehiculos = pd.read_csv("../data/hogares_vehiculos.csv")
vehiculos.head()
```

```python
frecuencias = vehiculos["vehiculos"].value_counts().sort_index()
probabilidades = frecuencias / len(vehiculos)
```

```python
# Crear una tabla de resultados
tabla = pd.DataFrame({
    "hogares": frecuencias,
    "probabilidad": probabilidades,
    "porcentaje": probabilidades * 100
})

print(tabla.round(2))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Creando la gráfica

```python
# Configurar el estilo
sns.set_theme(style="whitegrid")

# Crear la gráfica de barras
sns.countplot(data=vehiculos, x="vehiculos")

# Personalizar la gráfica
plt.title("Cantidad de vehículos por hogar")
plt.xlabel("Número de vehículos")
plt.ylabel("Número de hogares")

plt.show()
```
