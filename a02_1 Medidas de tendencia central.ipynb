# 📊 Medidas de tendencia central

## 💼 Salarios de empleados

📥 **Importamos la librería pandas** para trabajar con datos en forma de tabla (DataFrames).

```python
import pandas as pd
```

📂 **Cargamos los datos de salarios** desde un archivo CSV que contiene la información de los empleados.

```python
salarios = pd.read_csv("data/salarios_empleados.csv")
```

🔍 **Mostramos la tabla de salarios** para revisar los datos.

```python
salarios
```

### 📐 Obtener la media

📊 **Calculamos la media (promedio) de los salarios.**
La media representa el valor promedio de todos los datos. `numeric_only=True` ignora columnas no numéricas.

```python
salarios.mean(numeric_only=True)
```

📈 La media es `6933.333333`

### 📏 Obtener la mediana

🔸 **La mediana es el valor central** cuando los datos están ordenados. Es útil cuando hay valores extremos (muy altos o muy bajos).

```python
salarios.median(numeric_only=True)
```

📏 La mediana es `3650.0`

### 🔁 Obtener la moda

📌 **La moda es el valor que más se repite.** Si no hay un único valor más frecuente, se muestran todos los posibles.

```python
salarios.mode(numeric_only=True)
```

❗ No hay un valor único que se repita más veces. Se muestran todos los valores posibles.

## 🧑‍🏫 Asistencias de un grupo en una semana

📥 **Leemos los datos de asistencias** en la semana desde un archivo CSV.

```python
asistencias = pd.read_csv("data/asistencias_clases.csv")
asistencias
```

### 📐 Obtener la media

📊 **Calculamos el promedio de asistencias** del grupo a lo largo de la semana.

```python
asistencias.mean(numeric_only=True)
```

📈 El promedio de asistencias es de `5.4`

### 📏 Obtener la mediana

🔸 **La mediana** nos da el número de asistencias que está justo en el medio del grupo de datos ordenados.

```python
asistencias.median(numeric_only=True)
```

📏 La mediana es de `6.0`

### 🔁 Obtener la moda

📌 **Calculamos la moda** para ver cuál es el número de asistencias más común entre los estudiantes.

```python
asistencias.mode(numeric_only=True)
```

🔁 La moda es `6`
