# 📊 Filtrado de datos y graficaciones
En este notebook aprenderemos a filtrar datos y a graficar los resultados usando pandas 🐼 y sus funciones integradas para visualización 🎨.

## 📥 1. Importar la librería pandas
`pandas` nos permite trabajar con datos tabulares a través de estructuras llamadas **DataFrames**.

```python
import pandas as pd
```

## 📄 2. Cargar el archivo CSV del Titanic y visualizar los datos
Usamos `read_csv()` para leer el archivo `titanic_v5.csv`, que contiene información sobre los pasajeros del Titanic 🛳️.

```python
titanic = pd.read_csv("data/titanic_v5.csv")
titanic
```

## 👩‍🦰 3. Filtrar a las mujeres y contar cuántas sobrevivieron
Creamos una condición para filtrar solo a las pasajeras, y luego usamos `value_counts()` para contar las que sobrevivieron (`1`) y las que no (`0`).

```python
mujeres = titanic["sexo"] == "female"
titanic[mujeres]["sobrevivio"].value_counts()
```

## 🥧 4. Gráfica circular de mujeres que sobrevivieron
Representamos gráficamente en un **gráfico de pastel** la cantidad de mujeres que sobrevivieron vs. no sobrevivieron.

```python
titanic[mujeres]["sobrevivio"].value_counts().plot(kind="pie")
```

## 👨 5. Filtrar a los hombres y contar cuántos sobrevivieron
Aplicamos una condición similar, pero ahora para los pasajeros de sexo masculino. Y graficar

```python
hombres = titanic["sexo"] == "male"
titanic[hombres]["sobrevivio"].value_counts()
```

## 👶 6. Filtrar pasajeros menores de 20 años
Creamos una condición booleana para seleccionar a los pasajeros con edad menor a 20 años.

```python
menores_a_20 = titanic["edad"] < 20
titanic[menores_a_20]
```

## 🎓 7. Contar clase de pasajeros menores de 20 años
Contamos cuántos de los pasajeros menores de 20 años pertenecían a cada clase del Titanic (`1ra`, `2da`, `3ra`).

```python
titanic[menores_a_20]["clase_pasajero"].value_counts()
```

## 📊 8. Gráfica de barras por clase de pasajeros menores de 20 años
Usamos un gráfico de barras para visualizar cuántos menores de 20 viajaban en cada clase del Titanic.

```python
titanic[menores_a_20]["clase_pasajero"].value_counts().plot(kind="bar")
```
