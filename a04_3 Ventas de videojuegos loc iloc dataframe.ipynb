# 🎮 Pandas `loc` e `iloc` con DataFrames

Cuando trabajamos con un **DataFrame**, es muy común necesitar acceder a filas y columnas específicas. Para ello, **Pandas** ofrece dos herramientas muy poderosas: `loc` e `iloc`.

Aunque ambas permiten seleccionar datos, existe una diferencia importante: **`loc` utiliza etiquetas (índices y nombres de columnas)**, mientras que **`iloc` trabaja con posiciones numéricas**. Conocer cuándo utilizar cada una facilitará la consulta y manipulación de los datos.

En esta práctica utilizaremos un **conjunto de datos de ventas de videojuegos**, lo que permitirá aplicar estos conceptos en un escenario real de análisis de datos.

## 🎯 ¿Qué aprenderás?

En esta sección aprenderás a:

- 📌 Comprender la diferencia entre `loc` e `iloc`.
- 📌 Seleccionar filas utilizando etiquetas con `loc`.
- 📌 Seleccionar filas mediante posiciones numéricas con `iloc`.
- 📌 Acceder a columnas específicas o a combinaciones de filas y columnas.
- 📌 Extraer subconjuntos de un `DataFrame` para realizar análisis más específicos.

## 💡 Ejemplo de uso

Imagina un conjunto de datos con información sobre videojuegos:

| Juego | Plataforma | Género | Ventas (millones) |
|--------|------------|---------|-------------------|
| Minecraft | PC | Sandbox | 300 |
| Mario Kart 8 | Switch | Carreras | 75 |
| GTA V | Multiplataforma | Acción | 215 |
| FIFA 24 | Multiplataforma | Deportes | 25 |

Con **`loc`** podrías seleccionar el registro del videojuego **Minecraft** utilizando su etiqueta o índice, mientras que con **`iloc`** podrías obtener el primer registro simplemente indicando su posición (`0`). También es posible seleccionar varias filas y columnas para analizar únicamente la información que necesitas.

> 🚀 Dominar **`loc`** e **`iloc`** te permitirá navegar por cualquier `DataFrame` con facilidad, realizar consultas precisas y preparar los datos para análisis, visualizaciones o modelos de ciencia de datos.

📥 **Importamos la librería Pandas** para trabajar con DataFrames.

```python
import pandas as pd
```

📄 **Cargamos el archivo CSV** con datos de ventas de videojuegos.

```python
ventas = pd.read_csv("data/ventas_videojuegos.csv")
```

🔢 **Establecemos la columna `Titulo` como índice** para facilitar el uso de `loc`.

```python
ventas.set_index("Titulo", inplace=True)
ventas
```

## 🔍 Recuperar registros con `loc[ ]`
`loc` permite acceder a filas por su etiqueta (nombre).

🎯 **Obtenemos un solo registro por su título.**

```python
ventas.loc["Halo 3"]
```

📦 **Obtenemos un registro en formato DataFrame (doble corchete).**

```python
ventas.loc[["Halo 3"]]
```

🟩 **Buscamos otro videojuego por su título.**

```python
ventas.loc[["Minecraft"]]
```

🗃️ **Buscamos varios videojuegos al mismo tiempo.**

```python
ventas.loc[["Minecraft", "Halo 3", "Batman: Arkham City"]]
```

🔠 **Ordenamos el DataFrame por el índice (títulos).**

```python
ventas.sort_index(inplace=True)
ventas
```

🔍 **Seleccionamos un rango de títulos (inclusive).**

```python
ventas.loc["FIFA Soccer 13": "Kinect Sports"]
```

🧩 **Seleccionamos un rango de títulos y columnas específicas.**

```python
ventas.loc["Kinect Sports":"Resident Evil 5", ['Plataforma', 'Publisher']]
```

📊 **Seleccionamos columnas de género y ventas en Norteamérica.**

```python
ventas.loc["FIFA Soccer 13": "Kinect Sports", ["Genero", "Ventas_NA"]]
```

## 🔢 Recuperar registros con índice numérico usando `loc`
`loc` también se puede usar con índices numéricos si no hay un índice personalizado.

🔄 **Volvemos a cargar los datos para usar el índice numérico.**

```python
ventas = pd.read_csv("data/ventas_videojuegos.csv")
ventas
```

📌 **Seleccionamos registros específicos por su número de fila.**

```python
ventas.loc[[1, 6, 9]]
```

📚 **Seleccionamos un rango de filas continuas.**

```python
ventas.loc[5:9]
```

⏩ **Seleccionamos filas con un paso de 2 en el rango.**

```python
ventas.loc[5:9:2]
```

🔢 **Buscamos cuántas veces aparece un valor específico en una columna.**

```python
ventas["Ventas_NA"].value_counts().loc[14]
```

## 🎯 Recuperar registros con `iloc`
`iloc` se usa para acceder a registros por su posición numérica.

🧱 **Establecemos `Titulo` como índice de nuevo para futuras operaciones.**

```python
ventas.set_index("Titulo", inplace=True)
ventas
```

🔖 **Seleccionamos el primer registro con `iloc`.**

```python
ventas.iloc[[0]]
```

🎯 **Seleccionamos múltiples filas específicas por posición.**

```python
ventas.iloc[[1, 4, 5, 8]]
```

📏 **Seleccionamos un rango de filas por posición.**

```python
ventas.iloc[1:4]
```

🔃 **Ordenamos el DataFrame por el título.**

```python
ventas.sort_values("Titulo", inplace=True)
ventas
```

🔖 **Mostramos el primer registro luego de ordenar.**

```python
ventas.iloc[[0]]
```

📌 **Accedemos nuevamente a múltiples filas por posición.**

```python
ventas.iloc[[1, 4, 5, 8]]
```

🧮 **Usamos `iloc` con un rango de posiciones.**

```python
ventas.iloc[1:4]
```

🔄 **Volvemos a cargar los datos desde cero.**

```python
ventas = pd.read_csv("data/ventas_videojuegos.csv")
ventas
```

🔽 **Ordenamos el índice en orden descendente.**

```python
ventas.sort_index(ascending=False, inplace=True)
ventas
```

🎯 **Mostramos el primer registro después del orden descendente.**

```python
ventas.iloc[[0]]
```
