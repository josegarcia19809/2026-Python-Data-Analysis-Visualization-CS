## 📦 Importar librerías

Importamos la librería pandas, que nos permitirá trabajar con datos en forma de tablas (DataFrames) de manera eficiente.

```python
import pandas as pd
```

## 📄 Cargar el archivo CSV

Leemos el archivo CSV que contiene el reporte de felicidad mundial y lo guardamos en un DataFrame llamado `paises`. Luego mostramos el contenido para ver las primeras filas.

```python
paises = pd.read_csv("data/reporte_felicidad_mundial_2021.csv")
paises
```

## 🔢 Establecer el índice con el nombre del país

Configuramos el DataFrame para que el índice sea la columna 'Nombre del país', esto facilita buscar y organizar datos por país.

```python
paises.set_index("Nombre del país", inplace=True)
```

Mostramos el DataFrame para verificar que el índice fue cambiado correctamente.

```python
paises
```

## 💉 Ver la columna de esperanza de vida saludable

Extraemos y mostramos la columna que contiene la esperanza de vida saludable para todos los países.

```python
paises["Esperanza de vida saludable"]
```

## 📈 Graficar los primeros países según esperanza de vida saludable

Realizamos una gráfica simple con los primeros 5 países en la columna de esperanza de vida saludable para visualizar su comparación.

```python
paises["Esperanza de vida saludable"].head().plot()
```

## 🔄 Recargar el CSV estableciendo el índice directamente

Volvemos a cargar el archivo CSV, esta vez indicamos desde el inicio que la primera columna será el índice, evitando tener que establecerlo después.

```python
paises = pd.read_csv("data/reporte_felicidad_mundial_2021.csv", index_col=0)
paises
```

## 😊 Mostrar el puntaje de felicidad por país

Mostramos la columna del puntaje de felicidad para analizar los valores de cada país.

```python
paises["Puntaje de felicidad"]
```

## 🔤 Ordenar por índice (nombre del país)

Ordenamos el DataFrame alfabéticamente por el nombre del país (índice).

```python
paises.sort_index()
```

Ordenamos el DataFrame según el puntaje de felicidad de forma ascendente.

```python
paises.sort_values("Puntaje de felicidad")
```

Ordenamos el índice de los países en orden descendente y aplicamos el cambio al DataFrame.

```python
paises.sort_index(ascending=False, inplace=True)
```

Mostramos el DataFrame para verificar el nuevo orden.

```python
paises
```

## 🔍 Buscar datos de países específicos

Utilizamos `.loc` para buscar y mostrar datos de un país en específico por su nombre.

```python
paises.loc[["Yemen"]]
```

Mostramos los datos de Canadá.

```python
paises.loc[["Canada"]]
```

Mostramos los datos de varios países a la vez.

```python
paises.loc[["Canada", "Mexico", "United States"]]
```

## 📚 Seleccionar rango de países por índice

Primero ordenamos el índice para poder seleccionar un rango continuo de países.

```python
paises.sort_index(inplace=True)
```

Seleccionamos todos los países entre 'Denmark' y 'France', inclusive.

```python
paises.loc["Denmark":"France"]
```

## 🔢 Selección por posición con `.iloc`

Seleccionamos filas específicas por su posición numérica en el DataFrame.

```python
paises.iloc[[0]]
```

Seleccionamos varias filas en posiciones concretas.

```python
paises.iloc[[10,20,30,100]]
```

Seleccionamos un rango de filas por posición.

```python
paises.iloc[50:60]
```

## 🔎 Filtrar por columna usando `.loc`

Filtramos un rango de países y mostramos solo la columna 'Puntaje de felicidad' para ese rango.

```python
paises.loc["Canada": "Denmark", ['Puntaje de felicidad']]
```
