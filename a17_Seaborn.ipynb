# Visualizations With Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## load_dataset()
Seaborn comes with a handy way to quickly get some datasets to play with, but please note this is NOT the normal way of loading a CSV file.  Usually we'd use pandas.read_csv() as we've seen so far.

### 🍽️📊 Cargar el conjunto de datos `tips`

Esta operación utiliza la función `load_dataset()` de Seaborn para cargar el conjunto de datos llamado **`tips`**.

Este conjunto de datos contiene información sobre cuentas y propinas registradas en un restaurante, incluyendo variables como:

- 💵 `total_bill`: importe total de la cuenta.
- 💰 `tip`: propina otorgada.
- 🚻 `sex`: sexo del cliente.
- 🚬 `smoker`: indica si el cliente es fumador.
- 📅 `day`: día de la semana.
- ⏰ `time`: momento del día (almuerzo o cena).
- 👥 `size`: tamaño del grupo.

El DataFrame resultante se almacena en la variable `tips` y puede utilizarse para practicar análisis de datos, visualización y técnicas de manipulación con Pandas y Seaborn. 📈🍴

```python
tips = sns.load_dataset("tips")
```

## Seaborn Scatterplots

### 🎨 Configurar el estilo visual de Seaborn

Esta operación establece el tema visual predeterminado de Seaborn para los gráficos que se generen posteriormente.

Al aplicar el tema, Seaborn ajusta automáticamente aspectos como:

- 🎨 Estilo general del gráfico.
- 📝 Tipografía y tamaño de texto.
- 📏 Líneas de cuadrícula.
- 📊 Apariencia de ejes y etiquetas.

El objetivo es mejorar la legibilidad y la presentación de las visualizaciones sin necesidad de realizar configuraciones adicionales en cada gráfico. 📈✨

```python
sns.set_theme()
```

### 📈 Visualizar la relación entre el total de la cuenta y la propina

Esta operación crea un **gráfico de dispersión** (*scatter plot*) utilizando el conjunto de datos `tips`.

En el gráfico:

- 💵 `total_bill` se representa en el eje horizontal (X).
- 💰 `tip` se representa en el eje vertical (Y).

Cada punto corresponde a una cuenta registrada en el restaurante. La posición del punto muestra el importe total de la cuenta y la propina asociada.

Este tipo de gráfico es útil para identificar patrones, tendencias o posibles relaciones entre dos variables numéricas. Por ejemplo, permite observar si las propinas tienden a aumentar conforme aumenta el valor de la cuenta. 🍽️📊

```python
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

### 📈 Crear un gráfico de dispersión con Pandas

Esta operación genera un **gráfico de dispersión** (*scatter plot*) utilizando el método `plot()` de Pandas.

Se utiliza la columna `total_bill` como eje horizontal (X) y la columna `tip` como eje vertical (Y).

Cada punto representa una observación del conjunto de datos, mostrando la relación entre el importe total de una cuenta y la propina recibida.

Este gráfico permite visualizar posibles tendencias, correlaciones o patrones entre ambas variables, ayudando a responder preguntas como si las propinas aumentan cuando aumenta el valor de la cuenta. 🍽️💵📊

```python
tips.plot(kind="scatter", x="total_bill", y="tip")
```

### 🚬 `hue="smoker"`

El parámetro `hue` utiliza una variable categórica para diferenciar los datos mediante colores.

En este caso, los puntos del gráfico se colorean según el valor de la columna `smoker`, permitiendo distinguir visualmente entre clientes fumadores y no fumadores.

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="smoker")
```

### 🚻 `hue="sex"`

El parámetro `hue` colorea los puntos según los valores de la columna `sex`.

Esto permite distinguir visualmente los registros correspondientes a hombres y mujeres dentro del mismo gráfico.

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")
```

### 📅 `hue="day"`

El parámetro `hue` colorea los puntos según los valores de la columna `day`.

Cada día de la semana presente en el conjunto de datos se representa con un color diferente, lo que facilita comparar visualmente el comportamiento de las cuentas y propinas entre distintos días.

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="day")
```

 🚻 `hue="sex"`

Colorea los puntos según el sexo de los clientes, permitiendo diferenciar visualmente hombres y mujeres.


 ⏰ `style="time"`

El parámetro `style` utiliza diferentes formas o marcadores para representar los valores de la columna `time`.

Esto permite distinguir visualmente los registros correspondientes al almuerzo (`Lunch`) y la cena (`Dinner`) mediante símbolos distintos, además de los colores definidos por `hue`.

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex", style="time")
```

### 🚻 `style="sex"`

El parámetro `style` utiliza diferentes formas o marcadores según los valores de la columna `sex`.

Como también se está usando `hue="sex"`, la misma variable se representa mediante dos características visuales:

- 🎨 Color de los puntos.
- 🔷 Forma de los marcadores.

Esto hace más fácil distinguir los grupos, especialmente cuando el gráfico se imprime en blanco y negro o cuando los colores son difíciles de diferenciar.

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex", style="sex")
```

### 🔍 `size="size"`

El parámetro `size` modifica el tamaño de los puntos según los valores de la columna `size`.

Los registros con valores más altos en esta columna se muestran con marcadores más grandes, mientras que los valores más pequeños se representan con marcadores más pequeños.

Esto permite incorporar una tercera variable numérica al gráfico además de las variables representadas en los ejes X e Y. 📊

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", size="size")
```

### 🚻 `hue="sex"`

Colorea los puntos según el sexo de los clientes, permitiendo distinguir visualmente cada categoría.

### 🔍 `size="size"`

Ajusta el tamaño de los puntos de acuerdo con los valores de la columna `size`.

De esta forma, el gráfico representa cuatro variables al mismo tiempo:

- 📍 Posición horizontal: `total_bill`
- 📍 Posición vertical: `tip`
- 🎨 Color: `sex`
- 🔵 Tamaño del punto: `size`

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", size="size", hue="sex")
```
