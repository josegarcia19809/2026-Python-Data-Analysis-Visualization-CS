## Seaborn Scatterplots

### 🎨 Configurar el estilo visual de Seaborn

Esta operación establece el tema visual predeterminado de Seaborn para los gráficos que se generen posteriormente.

Al aplicar el tema, Seaborn ajusta automáticamente aspectos como:

- 🎨 Estilo general del gráfico.
- 📝 Tipografía y tamaño de texto.
- 📏 Líneas de cuadrícula.
- 📊 Apariencia de ejes y etiquetas.

El objetivo es mejorar la legibilidad y la presentación de las visualizaciones sin necesidad de realizar configuraciones adicionales en cada gráfico. 📈✨

```python
sns.set_theme()
```

### 📈 Visualizar la relación entre el total de la cuenta y la propina

Esta operación crea un **gráfico de dispersión** (*scatter plot*) utilizando el conjunto de datos `tips`.

En el gráfico:

- 💵 `total_bill` se representa en el eje horizontal (X).
- 💰 `tip` se representa en el eje vertical (Y).

Cada punto corresponde a una cuenta registrada en el restaurante. La posición del punto muestra el importe total de la cuenta y la propina asociada.

Este tipo de gráfico es útil para identificar patrones, tendencias o posibles relaciones entre dos variables numéricas. Por ejemplo, permite observar si las propinas tienden a aumentar conforme aumenta el valor de la cuenta. 🍽️📊

```python
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

### 📈 Crear un gráfico de dispersión con Pandas

Esta operación genera un **gráfico de dispersión** (*scatter plot*) utilizando el método `plot()` de Pandas.

Se utiliza la columna `total_bill` como eje horizontal (X) y la columna `tip` como eje vertical (Y).

Cada punto representa una observación del conjunto de datos, mostrando la relación entre el importe total de una cuenta y la propina recibida.

Este gráfico permite visualizar posibles tendencias, correlaciones o patrones entre ambas variables, ayudando a responder preguntas como si las propinas aumentan cuando aumenta el valor de la cuenta. 🍽️💵📊

```python
tips.plot(kind="scatter", x="total_bill", y="tip")
```

El parámetro `hue` colorea los puntos según los valores de la columna `sex`.

Esto permite distinguir visualmente los registros correspondientes a hombres y mujeres dentro del mismo gráfico.

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")
```

### 📅 `hue="day"`

El parámetro `hue` colorea los puntos según los valores de la columna `day`.

Cada día de la semana presente en el conjunto de datos se representa con un color diferente, lo que facilita comparar visualmente el comportamiento de las cuentas y propinas entre distintos días.

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="day")
```

## Seaborn Line Plots

### ✈️ Cargar el conjunto de datos `flights`

Carga el conjunto de datos `flights` de Seaborn y lo almacena en la variable `flights`.

Este conjunto contiene información sobre el número de pasajeros aéreos por mes y por año, siendo útil para practicar análisis de series temporales y gráficos de líneas. 📈

```python
flights = sns.load_dataset("flights")
```

### 📈 Gráfico de líneas

Representa la evolución de la variable `passengers` a lo largo de los años.

Como existen varios registros para cada año (uno por cada mes), Seaborn calcula automáticamente la **media** de pasajeros para cada año antes de dibujar la línea.

```python
sns.lineplot(data=flights, x="year", y="passengers")
```

### 📊 Agrupar y graficar

Agrupa los registros por año y calcula el promedio de pasajeros para cada grupo.

Posteriormente, el método `plot()` genera un gráfico de líneas utilizando esos promedios anuales.

```python
flights.groupby("year")["passengers"].mean().plot()
```

### ➕ `estimator="sum"`

Indica que los valores de pasajeros deben agregarse mediante una **suma** para cada año.

En lugar de mostrar el promedio anual, la línea representará el total de pasajeros registrados en cada año.

```python
sns.lineplot(data=flights, x="year", y="passengers", estimator="sum")
```

### 🎨 `hue="month"`

Crea una línea independiente para cada mes utilizando colores diferentes.

Esto permite comparar cómo cambia el número de pasajeros de cada mes a lo largo de los años.

```python
sns.lineplot(data=flights, x="year", y="passengers", hue="month")
```

### 📅 `parse_dates=["pickup", "dropoff"]`

Convierte automáticamente las columnas `pickup` y `dropoff` al tipo fecha y hora (`datetime`).

Esto permite realizar operaciones relacionadas con fechas, como extraer el año, mes, día, hora o calcular diferencias de tiempo.

```python
trips= sns.load_dataset("taxis", parse_dates=["pickup", "dropoff"])
```

### 🕒 `.dt.hour`

Extrae la hora de cada valor de la columna `pickup` y la almacena en una nueva columna llamada `hour`.

Los valores resultantes estarán en un rango de **0 a 23**, representando las horas del día.

```python
trips["hour"] = trips["pickup"].dt.hour
```

```python
trips
```

### 🎨 `hue="payment"`

Crea una línea independiente para cada método de pago utilizando colores diferentes.

### 🔷 `style="color"`

Utiliza diferentes estilos de línea según los valores de la columna `color`.

### 🚫 `errorbar=None`

Oculta las bandas de error que Seaborn muestra por defecto alrededor de las líneas, dejando visible únicamente la línea principal.

```python
sns.lineplot(data=trips, x="hour", y="total", hue="payment", style="color", errorbar=None)
```

## The Figure-Level Relplot( ) Method

```python
tips
```

### 🔍 `kind="scatter"`

Indica que se debe crear un gráfico de dispersión (*scatter plot*).

### 🚻 `col="sex"`

Divide la visualización en varias columnas, creando un gráfico independiente para cada valor de la columna `sex`.

Esto permite comparar los datos de hombres y mujeres en gráficos separados.

```python
sns.relplot(data=tips, x="total_bill", y="tip", kind="scatter", col="sex")
```

### 🚬 `hue="smoker"`

Colorea los puntos según si el cliente es fumador o no.

### 🚻 `col="sex"`

Crea una columna de gráficos para cada valor de la columna `sex`, permitiendo comparar hombres y mujeres por separado.

```python
sns.relplot(data=tips, x="total_bill", y="tip", kind="scatter", hue="smoker", col="sex")
```

### 🚬 `hue="smoker"`

Colorea los puntos según si el cliente es fumador o no.

### 🚻 `col="sex"`

Crea una columna de gráficos para cada valor de la columna `sex`.

### ⏰ `row="time"`

Crea una fila de gráficos para cada valor de la columna `time`.

El resultado es una cuadrícula de gráficos donde cada combinación de sexo y momento del día (almuerzo o cena) se muestra en un panel independiente, facilitando las comparaciones entre grupos.

```python
sns.relplot(data=tips, x="total_bill", y="tip", kind="scatter", hue="smoker", col="sex", row="time")
```

### 📈 `kind="line"`

Indica que se deben crear gráficos de líneas.

### 🏙️ `col="pickup_borough"`

Genera una columna de gráficos para cada distrito de recogida (*borough*).

Cada distrito se muestra en un panel independiente.

### 🎨 `hue="payment"`

Dibuja una línea diferente para cada método de pago utilizando colores distintos dentro de cada gráfico.

```python
sns.relplot(data=trips, x="hour", y="total", kind="line", col="pickup_borough", hue="payment")
```

### 🏙️ `col="pickup_borough"`

Crea una columna de gráficos para cada distrito de recogida (*pickup borough*).

### 📍 `row="dropoff_borough"`

Crea una fila de gráficos para cada distrito de destino (*dropoff borough*).

### 🎨 `hue="payment"`

Dibuja una línea diferente para cada método de pago utilizando colores distintos dentro de cada gráfico.

El resultado es una cuadrícula de gráficos donde cada panel representa una combinación específica de distrito de recogida y distrito de destino.

```python
sns.relplot(
    data=trips, 
    x="hour",
    y="total", 
    kind="line", 
    col="pickup_borough", 
    hue="payment", 
    row="dropoff_borough"
)
```

Colorea las barras del histograma según los valores de la columna `time`.

### 🚬 `hue="smoker"`

Apila las barras correspondientes a cada categoría una encima de otra.

### 🚬 `hue="smoker"`

Diferencia las observaciones según si el cliente es fumador o no mediante colores distintos.

Muestra las barras de cada categoría una al lado de la otra dentro de cada intervalo.

```python
sns.histplot(data=tips, x="tip", hue="smoker", multiple="dodge")
```

En este caso, los datos se agrupan en rangos de **500 gramos**, por ejemplo:

Esto permite controlar el nivel de detalle con el que se muestra la distribución de los datos.

Apila las categorías dentro de cada intervalo del histograma.

### ↕️ `y="tip"`

Dibuja las marcas (*rugs*) sobre el eje vertical utilizando los valores de la columna `tip`.

### 📏 `height=0.2`

Define la longitud de las marcas en relación con el tamaño del gráfico.

```python
sns.rugplot(data=tips, y="tip", height=0.2)
```

### 📈 `sns.kdeplot(...)`

Genera una curva de densidad suavizada para mostrar la distribución de los valores de `total_bill`.

### 📏 `height=0.07`

### 🔹 `s=5`

Controla el tamaño de los puntos del gráfico de dispersión.

Un valor de `5` genera marcadores muy pequeños, lo cual es útil cuando existen miles de observaciones y se desea evitar el exceso de superposición.

### 📏 `lw=1`

Define el grosor (*line width*) de las marcas del gráfico *rug*.

### 🌫️ `alpha=0.005`

Controla la transparencia de las marcas.

Un valor muy pequeño como `0.005` hace que cada marca sea casi invisible de forma individual, pero las zonas con muchas observaciones se vuelven más visibles debido a la acumulación de marcas.

Esta técnica ayuda a identificar regiones de alta densidad cuando se trabaja con conjuntos de datos grandes.

```python
diamonds = sns.load_dataset("diamonds")
sns.scatterplot(data=diamonds, x="carat", y="price", s=5)
sns.rugplot(data=diamonds, x="carat", y="price", lw=1, alpha=.005)
```

## The Figure-Level displot( ) Method

### 📊 `kind="hist"`

Indica que se debe crear un histograma.

### 📏 `height=3`

Define la altura del gráfico en pulgadas.

### ↔️ `aspect=2`

Define la relación entre ancho y alto.

Un valor de `2` hace que el gráfico sea aproximadamente dos veces más ancho que alto.

```python
sns.displot(kind="hist", data=penguins, x="body_mass_g", height=3, aspect=2)
```

### 🚻 `hue="sex"`

Diferencia los datos según el sexo utilizando colores distintos.

### 🐧 `col="species"`

Crea un histograma independiente para cada especie de pingüino, mostrándolos en columnas separadas.

```python
sns.displot(
    kind="hist", 
    data=penguins, 
    hue="sex",
    x="body_mass_g", 
    col="species"
)
```

### 📈 `kind="kde"`

Indica que se debe crear un gráfico de densidad KDE en lugar de un histograma.

### ⏰ `col="time"`

Genera un gráfico independiente para cada valor de la columna `time`.

### 📏 `rug=True`

Añade marcas (*rug plots*) sobre el eje para mostrar la posición exacta de cada observación utilizada para construir las curvas de densidad.

```python
sns.displot(data=tips, kind="kde", x="tip", col="time", rug=True)
```

### 📈 `kind="kde"`

Indica que se debe crear un gráfico de densidad KDE.

### ↔️ `y="tip"`

Al especificar una variable en `y`, Seaborn genera una **KDE bidimensional**, mostrando la densidad conjunta de ambas variables mediante curvas de nivel.

### 📏 `rug=True`

Añade marcas (*rug plots*) sobre los ejes para indicar la posición exacta de las observaciones en los datos.

```python
sns.displot(data=tips, kind="kde", x="total_bill", y="tip", rug=True)
```
