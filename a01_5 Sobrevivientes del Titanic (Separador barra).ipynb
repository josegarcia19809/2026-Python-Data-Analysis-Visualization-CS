## `read_csv` con separadores sin coma

Cuando un archivo CSV usa un separador diferente a la coma, como una barra vertical (`|`), al cargarlo sin especificar el separador, Pandas interpreta todo como una sola columna.

Para que Pandas lea correctamente cada columna, debemos indicar explícitamente el separador usando el parámetro `sep`.

```python
import pandas as pd
```

Primero, veamos qué pasa si cargamos el archivo sin especificar el separador:

```python
titanic = pd.read_csv("data/titanic_v4.csv")
titanic
```

Como podemos ver, todo queda en una sola columna porque el archivo está separado por `|`.

Ahora, le indicamos a Pandas que use `|` como separador para leer correctamente las columnas:

```python
titanic = pd.read_csv("data/titanic_v4.csv", sep="|")
titanic
```
