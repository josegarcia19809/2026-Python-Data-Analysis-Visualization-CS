# 📊 Series: Usando `value_counts()` y `plot()` para visualizar datos

Una de las formas más rápidas de comprender un conjunto de datos es **contar cuántas veces aparece cada valor** y representar esa información mediante una gráfica. En **Pandas**, esto se logra fácilmente utilizando los métodos `value_counts()` y `plot()` sobre una **Series**.

El método `value_counts()` calcula la frecuencia de cada valor único de una columna, mientras que `plot()` permite convertir esos resultados en gráficos que facilitan la interpretación de los datos. Esta combinación es muy útil para explorar variables categóricas, identificar tendencias y comunicar resultados de manera visual.

## 🎯 ¿Qué aprenderás?

En esta sección aprenderás a:

- 📌 Contar la frecuencia de los valores de una columna con `value_counts()`.
- 📌 Identificar las categorías más y menos frecuentes.
- 📌 Crear gráficas directamente desde una **Series** utilizando `plot()`.
- 📌 Elegir el tipo de gráfico más adecuado según los datos.
- 📌 Interpretar los resultados obtenidos para apoyar el análisis de datos.

## 💡 Ejemplo de uso

Imagina un conjunto de datos con la carrera de varios estudiantes:

| Carrera |
|----------|
| Informática |
| Civil |
| Informática |
| Química |
| Civil |
| Informática |

Al aplicar `value_counts()`, obtendrás el número de estudiantes inscritos en cada carrera. Posteriormente, con `plot()`, podrás visualizar esa información mediante una gráfica de barras o una gráfica de pastel, facilitando la comparación entre categorías.

> 🚀 La combinación de **`value_counts()`** y **`plot()`** es una herramienta esencial en el análisis exploratorio de datos (EDA), ya que permite transformar datos en información visual de forma rápida y con muy pocas líneas de código.

## 📁 Conjunto de datos del Titanic
Analizaremos columnas usando `value_counts()` para ver cuántas veces aparece cada valor 🔢 y después visualizaremos con `plot()` 📈.

```python
import pandas as pd
```

📥 Cargamos los datos del archivo CSV y los almacenamos en el DataFrame `titanic`.

```python
titanic = pd.read_csv("data/titanic_v5.csv")
titanic
```

### 🔍 Mostramos los valores de la columna `edad` para observar qué contiene

```python
titanic["edad"]
```

### 🔢 Frecuencia de aparición de valores en columnas
Vamos a contar cuántas veces aparece cada valor en diferentes columnas del DataFrame.

```python
# Frecuencia de cada edad 👶👵
titanic["edad"].value_counts()
```

```python
# Frecuencia por sexo 🚹🚺
titanic["sexo"].value_counts()
```

```python
# Frecuencia de sobrevivientes (1) y no sobrevivientes (0) ⚰️🚢
titanic["sobrevivio"].value_counts()
```

```python
# Total de sobrevivientes sumando los 1️⃣
titanic["sobrevivio"].sum()
```

```python
# Frecuencia por clase del pasajero 🥇🥈🥉
titanic["clase_pasajero"].value_counts()
```

```python
# Clase con más pasajeros 👑
titanic["clase_pasajero"].value_counts().head(1)
```

```python
# Sexo más frecuente entre los pasajeros 🧑‍🤝‍🧑
titanic["sexo"].value_counts().head(1)
```

## 📉 Sacando los porcentajes por categoría
`normalize=True` nos da proporciones, las multiplicamos por 100 para convertirlas en porcentajes 📊.

```python
titanic["clase_pasajero"].value_counts(normalize=True) * 100
```

## 🧮 Mostrando solo las primeras frecuencias de edad

```python
titanic["edad"].value_counts().head()
```

### 🔽 Ordenando frecuencias en orden ascendente
Ideal para ver qué edades se repiten menos.

```python
titanic["edad"].value_counts(ascending=True)
```

### 👯‍♂️ Combinación de columnas: `sexo` y `edad`
Nos permite saber combinaciones exactas y su frecuencia.

```python
titanic[["sexo", "edad"]].value_counts()
```

## 📊 Gráfica de barras para `edad`
Visualizamos la frecuencia de cada edad como barras verticales 📈.

```python
titanic["edad"].value_counts().plot(kind="bar")
```

## 🥧 Gráfica de pastel para `sexo`
Ideal para comparar proporciones de manera visual.

```python
titanic["sexo"].value_counts().plot(kind="pie")
```

## 🥧 Gráfica de pastel para `sobrevivio`
Muestra visualmente cuántos pasajeros sobrevivieron frente a los que no.

```python
titanic["sobrevivio"].value_counts().plot(kind="pie")
```

## 🥧 Gráfica de pastel para `clase_pasajero`
Para ver cuántos pasajeros había por clase de forma visual.

```python
titanic["clase_pasajero"].value_counts().plot(kind="pie")
```

## 📊 Gráfica de barras para `clase_pasajero`
Una forma clara de ver qué clase tuvo más pasajeros o sobrevivientes.

```python
titanic["clase_pasajero"].value_counts().plot(kind="bar")
```
