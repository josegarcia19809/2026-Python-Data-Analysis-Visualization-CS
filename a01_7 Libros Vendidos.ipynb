# 📚 Análisis de Libros Vendidos
Este cuaderno realiza un análisis estadístico básico sobre un conjunto de datos de libros vendidos. 📊

```python
import pandas as pd
```

## 🗂️ Cargar el conjunto de datos

```python
libros = pd.read_csv("data/libros_vendidos.csv")
libros
```

## 📈 Máximos por columna
Muestra el valor máximo de cada columna (texto o número).

```python
libros.max()
```

## 📉 Mínimos por columna
Muestra el valor mínimo de cada columna.

```python
libros.min()
```

## ➕ Suma total (solo numéricos)
Suma los valores numéricos de las columnas del DataFrame.

```python
libros.sum(numeric_only=True)
```

## 🔍 Selección de filas específicas
Muestra las 2 primeras de las últimas 4 filas.

```python
libros.tail(4).head(2)
```

## ➕ Suma de una porción del DataFrame
Suma de columnas numéricas de esas 2 filas seleccionadas.

```python
libros.tail(4).head(2).sum(numeric_only=True)
```

## 🔢 Conteo de valores no nulos
Muestra cuántos valores hay por columna (útil para ver si hay datos perdidos).

```python
libros.count()
```

## 📊 Promedios (media)
Calcula el promedio de cada columna numérica.

```python
libros.mean(numeric_only=True)
```

## 🔸 Mediana
Valor central de cada columna numérica.

```python
libros.median(numeric_only=True)
```

## 🔁 Moda
Valor más frecuente en cada columna numérica.

```python
libros.mode(numeric_only=True)
```

## 📋 Resumen estadístico general
Incluye conteo, media, desviación estándar, valores mínimo, máximo y percentiles.

```python
libros.describe()
```

## ℹ️ Información del DataFrame
Muestra el tipo de dato de cada columna y la cantidad de datos no nulos.

```python
libros.info()
```

## 🔠 Estadísticas para columnas de texto
Usamos `include='object'` para ver estadísticas sobre columnas no numéricas.

```python
libros.describe(include='object')
```
