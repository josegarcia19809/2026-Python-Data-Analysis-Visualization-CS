# 📺 Filtrado de datos usando el operador OR (|). Títulos de Netflix

📥 **Importamos la librería `pandas`** para poder trabajar con datos en forma de tabla (DataFrames).

```python
import pandas as pd
```

📄 **Leemos el archivo CSV** que contiene los títulos disponibles en Netflix.

```python
netflix = pd.read_csv("data/titulos_netflix.csv")
```

🔍 **Visualizamos los últimos registros** del DataFrame para conocer su estructura y contenido.

```python
netflix.tail()
```

🎬 **Filtramos los títulos que son películas o que tienen clasificación PG-13** usando el operador `|` (OR).

```python
# Mostrar los programas de Netflix que sean películas o cuya clasificación sea PG-13
peliculas = netflix["tipo"] == "Movie"
clasificacion_pg_13 = netflix["clasificación"] == "PG-13"
netflix[peliculas | clasificacion_pg_13]
```

📅 **Filtramos los mismos programas anteriores pero ahora los ordenamos por año de lanzamiento** para ver cómo han cambiado con el tiempo.

```python
peliculas = netflix["tipo"] == "Movie"
clasificacion_pg_13 = netflix["clasificación"] == "PG-13"
programas = netflix[peliculas | clasificacion_pg_13]
programas.sort_values("año_lanzamiento")
```

🎥 **Filtramos los programas dirigidos por David Fincher o Martin Scorsese**, dos directores muy reconocidos.

```python
david = netflix["director"] == "David Fincher"
martin = netflix["director"] == "Martin Scorsese"
netflix[ david | martin]
```

🔤 **Mostramos los mismos programas pero ahora ordenados por título** para una mejor organización alfabética.

```python
david = netflix["director"] == "David Fincher"
martin = netflix["director"] == "Martin Scorsese"
netflix[ david | martin].sort_values("título")
```

📆 **Filtramos los programas dirigidos por Fincher o Scorsese que fueron lanzados después de 2015**. ¡Veamos sus obras más recientes!

```python
david = netflix["director"] == "David Fincher"
martin = netflix["director"] == "Martin Scorsese"
anio_lanzamiento = netflix["año_lanzamiento"] > 2015
netflix[ (david | martin) & anio_lanzamiento ]
```

✅ **¡Listo! Hemos practicado cómo filtrar datos usando el operador OR (`|`) en Pandas con ejemplos de títulos de Netflix.**
