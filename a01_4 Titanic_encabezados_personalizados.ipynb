## 📦 Importar pandas

Primero, importamos la biblioteca `pandas`, que nos permite trabajar con estructuras de datos tabulares como los archivos `.csv`.

````python
import pandas as pd
````

## 👁️ Visualizar el contenido original del archivo CSV

Cargamos el archivo sin modificar los encabezados para ver cómo vienen los nombres de columna originales.

````python
titanic = pd.read_csv("data/titanic_v3.csv")
titanic.head()
````

## 🏷️ Reemplazar los nombres de columnas con una lista personalizada

Si queremos usar nuestros propios nombres de columnas, podemos pasar una lista con ellos usando el parámetro `names`. Pero cuidado:
- Si no se especifica `header=0`, Pandas tratará la **primera fila de datos** como una **fila más**, no como encabezado.
- Esto puede duplicar el encabezado o alterar los datos.

👉 Primero veamos qué ocurre si no usamos `header=0`.

````python
nombres_columnas = ['CLASE PASAJERO', 'SOBREVIVIENTE', 'NOMBRE', 'SEXO', 'EDAD']
titanic = pd.read_csv("data/titanic_v3.csv", names=nombres_columnas)
titanic.head()
````

## ✅ Reemplazar nombres correctamente con `header=0`

Ahora sí, indicamos explícitamente que el archivo ya tiene encabezados en la primera fila, usando `header=0`.
De esta forma, Pandas ignorará los encabezados originales y usará nuestra lista `nombres_columnas`.

````python
nombres_columnas = ['CLASE PASAJERO', 'SOBREVIVIENTE', 'NOMBRE', 'SEXO', 'EDAD']
titanic = pd.read_csv("data/titanic_v3.csv", names=nombres_columnas, header=0)
titanic.head()
````

## ➕ Suma de columnas numéricas

Usamos `titanic.sum(numeric_only=True)` para sumar los valores de todas las columnas **numéricas** del DataFrame.  
Esto ignora automáticamente columnas con texto u objetos.

```python
titanic.sum(numeric_only=True)


````python
titanic.sum(numeric_only=True)
````


## 🔢 Conteo de valores no nulos

El método `titanic.count()` cuenta cuántos valores **no nulos (non-null)** hay en cada columna.  
Permite detectar si faltan datos en alguna de ellas.

```python
titanic.count()


````python
titanic.count()
````

## 📊 Estadísticas descriptivas de columnas numéricas

La función `titanic.describe()` muestra estadísticas generales (como media, desviación estándar, mínimo, máximo y percentiles)  
de todas las columnas **numéricas** del DataFrame.

```python
titanic.describe()

````python
titanic.describe()
````

## 📝 Estadísticas descriptivas de columnas tipo texto (object)

La instrucción `titanic.describe(include=["object"])` muestra estadísticas como cantidad de valores, valor más frecuente (top)  
y cuántas veces aparece (freq), pero solo para las columnas **de tipo texto (object)**.

```python
titanic.describe(include=["object"])

````python
titanic.describe(include=["object"])
````
