# 📋 Series: Selección de una sola columna

Cuando trabajamos con **Pandas**, una de las tareas más comunes es acceder a la información de una sola columna de un conjunto de datos. Al seleccionar una columna, **Pandas devuelve un objeto llamado `Series`**, una estructura de datos unidimensional que almacena valores junto con su índice.

Las **Series** son ideales para realizar operaciones de análisis, filtrado, cálculos estadísticos y transformaciones sobre una sola variable del conjunto de datos. Además, muchas de las funciones más utilizadas de Pandas trabajan directamente sobre este tipo de estructura.

## 🎯 ¿Qué aprenderás?

En esta sección aprenderás a:

- 📌 Seleccionar una columna de un `DataFrame`.
- 📌 Comprender qué es una **Series** y cómo se diferencia de un `DataFrame`.
- 📌 Acceder a los valores y al índice de una Series.
- 📌 Aplicar operaciones básicas sobre una columna.
- 📌 Utilizar métodos comunes para explorar y analizar los datos.

## 💡 Ejemplo de uso

Imagina un conjunto de datos de estudiantes con las columnas:

| Nombre | Edad | Carrera |
|--------|------|----------|
| Ana | 20 | Informática |
| Luis | 21 | Civil |
| Sofía | 19 | Química |

Si únicamente seleccionamos la columna **Edad**, el resultado será una **Series** que contiene únicamente las edades de los estudiantes, manteniendo el índice correspondiente a cada registro.

> 🚀 Dominar las **Series** es el primer paso para realizar análisis de datos eficientes con Pandas, ya que la mayoría de las operaciones comienzan seleccionando una o varias columnas de un conjunto de datos.

## 📁 Conjunto de datos del Titanic

```python
import pandas as pd
```

🔽 Cargando el conjunto de datos Titanic en un DataFrame para su análisis.

```python
titanic = pd.read_csv("data/titanic_v5.csv")
titanic
```

## 👤 Mostrando la columna `nombre` mediante la notación punto

```python
titanic.nombre
```

## 🔎 Mostrando la columna `nombre` mediante la notación corchetes
### ✅ Esta notación es más flexible y recomendada.

```python
titanic["nombre"]
```

## 🧠 Usar una variable para seleccionar una columna dinámicamente

```python
nombre_columna = "edad"
titanic[nombre_columna]
```

```python
# 🧬 Verificamos el tipo: pandas.Series
type(titanic[nombre_columna])
```

## ➕ Sumar los valores de una Series booleana como 'sobrevivio'

```python
titanic["sobrevivio"].sum()
```

## 🔝 Obtener el valor máximo de la columna `edad`

```python
titanic["edad"].max()
```

## 💾 Guardar una Series en una variable para manipularla después

```python
nombres = titanic["nombre"]
```

```python
nombres.values
```

## 👀 Obtener algunas filas del inicio y final de la Series

```python
titanic["nombre"].head()
```

```python
titanic["nombre"].tail()
```

## 📊 Estadísticas descriptivas de la columna `edad`

```python
titanic["edad"].describe()
```

## 🔁 Valores únicos que aparecen en columnas categóricas o numéricas

```python
titanic["clase_pasajero"].unique()
```

```python
titanic["edad"].unique()
```

```python
titanic["sexo"].unique()
```

## 🔢 Contar cuántos valores distintos hay por columna

```python
titanic["clase_pasajero"].nunique()
```

```python
titanic["edad"].nunique()
```

```python
titanic["sexo"].nunique()
```

## 📈 Obtener los valores extremos (grandes y chicos) de `edad`

```python
titanic["edad"].nlargest()
```

```python
titanic["edad"].nsmallest()
```

### 👶 Edad más baja registrada

```python
titanic["edad"].min()
```

### 👴 Edad más alta registrada

```python
titanic["edad"].max()
```

### 🔸 Obtener la mediana (valor medio) de la columna `edad`

```python
titanic["edad"].median()
```

### 🧮 Calcular la media solo de los primeros 5 registros

```python
titanic["edad"].head().mean()
```

```python
titanic["edad"].head()
```

## 🔍 Mostrar los registros con mayores edades del dataset

```python
titanic.nlargest(10, ["edad"])
```

## 🧓 Mostrar los 10 pasajeros más jóvenes

```python
titanic.nsmallest(10, ["edad"])
```
