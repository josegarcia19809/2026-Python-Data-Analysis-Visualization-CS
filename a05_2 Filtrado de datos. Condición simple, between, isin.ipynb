# 🎬 Filtrado de datos: Condición simple, `between`, `isin`
Exploraremos cómo filtrar datos en un DataFrame de Netflix usando condiciones simples, rangos de años (`between`) y listas de valores (`isin`).

📥 **Importamos la librería `pandas`** para trabajar con DataFrames.

```python
import pandas as pd
```

📄 **Cargamos el archivo CSV** que contiene los títulos de Netflix.

```python
netflix = pd.read_csv("data/titulos_netflix.csv")
```

👀 **Mostramos todo el DataFrame** para conocer su estructura y columnas disponibles.

```python
netflix
```

📅 **Filtramos las películas que se lanzaron después de 2018** usando una condición simple (`>`).

```python
netflix[netflix["año_lanzamiento"] > 2018]
```

🎥 **Películas lanzadas en 2020 o después** usando la condición `>=`.

```python
netflix[netflix["año_lanzamiento"] >= 2020]
```

📼 **Películas más antiguas**: lanzadas **antes del año 2000**.

```python
netflix[netflix["año_lanzamiento"] < 2000]
```

📆 **Películas entre 2000 y 2005**, incluyendo ambos extremos. Utilizamos `.between()` para mayor legibilidad.

```python
netflix[netflix["año_lanzamiento"].between(2000, 2005)]
```

🎬 **Filtramos las películas dirigidas por ciertos directores** usando `isin()` con una lista de nombres.

```python
directores =["Alfonso Cuarón","Tim Burton", "Steven Spielberg"]
netflix[netflix['director'].isin(directores)]
```

📊 **Contamos cuántas películas tiene cada uno de estos directores**.

```python
netflix[netflix['director'].isin(directores)]['director'].value_counts()
```

🔞 **Filtramos películas con clasificación para adultos**: "TV-MA", "R" y "NR".

```python
adultos = netflix["clasificación"].isin(["TV-MA", "R", "NR"])
netflix[adultos]
```

📈 **Contamos cuántas películas hay por cada una de estas clasificaciones para adultos**.

```python
netflix[adultos]['clasificación'].value_counts()
```
