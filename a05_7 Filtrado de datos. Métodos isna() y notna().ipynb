# 🎬 Filtrado de datos: Métodos `isna()` y `notna()`

Vamos a aprender cómo detectar y filtrar datos faltantes en un DataFrame de títulos de Netflix usando `isna()` y `notna()` de la librería **pandas**. 🧠📊

📥 **Paso 1: Importar la librería `pandas`**

Necesitamos `pandas` para manipular el DataFrame.

```python
import pandas as pd
```

📄 **Paso 2: Cargar el archivo CSV**

Leemos el archivo `titulos_netflix.csv` que contiene información como título, director y clasificación.

```python
netflix = pd.read_csv("data/titulos_netflix.csv")
```

🔍 **Paso 3: Ver la estructura del DataFrame**

Con `info()` vemos cuántos datos hay por columna y si existen valores nulos (missing values).

```python
netflix.info()
```

🤔 **Paso 4: Detectar valores nulos en la columna `director`**

`isna()` devuelve `True` donde falta el valor (es decir, donde hay `NaN`).

```python
netflix["director"].isna()
```

📌 **Paso 5: Filtrar filas donde falta el director**

Podemos ver solo los registros donde `director` no tiene datos.

```python
netflix[ netflix["director"].isna() ]
```

🔎 **Paso 6: Filtro doble: director y clasificación faltantes**

También podemos combinar condiciones usando `&` (AND lógico).

```python
netflix[ netflix["director"].isna() & netflix["clasificación"].isna() ]
```

✅ **Paso 7: Detectar valores *no nulos***

Con `notna()` podemos ver qué registros **sí** tienen director.

```python
netflix["director"].notna()
```

📋 **Paso 8: Mostrar los títulos que sí tienen director**

```python
netflix[ netflix["director"].notna() ]
```
