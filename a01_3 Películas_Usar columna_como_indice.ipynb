# 🎬 Calificaciones de películas (IMDb)

Este notebook trabaja con un conjunto de datos de películas provenientes de IMDb. Aquí aprenderás cómo:
- Leer un archivo CSV.
- Visualizar los primeros registros.
- Usar una columna como índice del DataFrame para mejorar la organización de los datos.

## 🧪 Paso 1: Importar la biblioteca Pandas

Primero, importamos la biblioteca `pandas`, que nos permite trabajar con datos tabulares como los que tiene IMDb.

````python
import pandas as pd
````

## 📥 Paso 2: Cargar el archivo CSV

Cargamos el archivo `imdb.csv` usando `pd.read_csv()`. Este archivo contiene información sobre películas y sus calificaciones.

````python
peliculas = pd.read_csv("data/imdb.csv")
````

## 👁️ Paso 3: Visualizar las primeras filas

Usamos `head()` para ver los primeros registros del DataFrame y así entender cómo están estructurados los datos.

````python
peliculas.head()
````

## 🔁 Paso 4: Usar una columna como índice

El archivo ya incluye una columna que podría usarse como índice (por ejemplo, un ID único o posición). Para indicarle a Pandas que esa columna debe ser el índice, usamos el parámetro `index_col=0`.

````python
peliculas = pd.read_csv("data/imdb.csv", index_col=0)
````

## 📌 Paso 5: Confirmar el nuevo índice

Volvemos a usar `head()` para comprobar que la primera columna ahora forma parte del índice del DataFrame.

````python
peliculas.head()
````

Esto nos ayuda a identificar si alguna columna tiene datos faltantes.
