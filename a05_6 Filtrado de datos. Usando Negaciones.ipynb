# ❗ Filtrado de datos usando negaciones
Aprenderemos a usar **negaciones (~)** para filtrar registros en un DataFrame utilizando pandas 🐼.

## 📥 1. Importar la librería pandas
Usamos la biblioteca `pandas` para manejar datos en forma de tablas (DataFrames) 📊.

```python
import pandas as pd
```

## 📄 2. Cargar el archivo CSV del Titanic
Leemos el archivo `titanic_v5.csv` que contiene información sobre los pasajeros del Titanic 🛳️.

```python
titanic = pd.read_csv("data/titanic_v5.csv")
```

## 👀 3. Visualizar el DataFrame completo
Mostremos una vista general del conjunto de datos cargado 🧾.

```python
titanic
```

## 🙋‍♀️ 4. Visualizar las primeras 5 personas
Usamos `head()` para ver las primeras filas del DataFrame y enfocarnos en un subconjunto de datos.

```python
primeras_5_personas = titanic.head()
primeras_5_personas
```

## ✅ 5. Crear una serie booleana para identificar mujeres
Creamos una condición que evalúa si la columna `sexo` es igual a `'female'`. Devuelve una serie booleana de `True` o `False`.

```python
mujeres = primeras_5_personas["sexo"] == "female"
mujeres
```

## 🔄 6. Crear la negación con `~`
Usamos `~` para negar la serie booleana anterior, invirtiendo `True` por `False` y viceversa.

```python
~mujeres
```

## 👨 7. Filtrar personas que **no sean mujeres**
Aplicamos la negación para mostrar a quienes **no** sean de sexo femenino (es decir, los hombres).

```python
primeras_5_personas[~mujeres]
```

## ❌ 8. Filtrar pasajeros que **no sobrevivieron** y **no eran de primera clase**
Combinamos dos condiciones negadas:
- Que **no sobrevivieron** (`~sobrevivieron`)
- Que **no viajaban en primera clase** (`~primera_clase`)

Usamos el operador `&` para aplicar ambas condiciones al mismo tiempo.

```python
# Condición: pasajeros que sobrevivieron
sobrevivieron = titanic["sobrevivio"] == 1

# Condición: pasajeros de primera clase
primera_clase = titanic["clase_pasajero"] == 1

# Negación de ambas condiciones combinadas
titanic[ ~sobrevivieron & ~primera_clase ]
```
