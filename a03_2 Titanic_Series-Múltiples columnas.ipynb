# 📋 DataFrame: Selección de múltiples columnas

En muchas ocasiones no basta con trabajar con una sola columna; es necesario analizar varias variables al mismo tiempo. En **Pandas**, cuando seleccionamos **dos o más columnas** de un `DataFrame`, el resultado sigue siendo un **DataFrame**, ya que conserva la estructura de filas y columnas.

Seleccionar múltiples columnas permite comparar información, crear subconjuntos de datos y preparar conjuntos de datos para análisis, visualizaciones o modelos de aprendizaje automático.

## 🎯 ¿Qué aprenderás?

En esta sección aprenderás a:

- 📌 Seleccionar dos o más columnas de un `DataFrame`.
- 📌 Comprender la diferencia entre obtener una **Series** y un **DataFrame**.
- 📌 Crear subconjuntos de datos con las columnas de interés.
- 📌 Cambiar el orden de las columnas seleccionadas.
- 📌 Preparar datos para su análisis o exportación.

## 💡 Ejemplo de uso

Imagina un conjunto de datos de estudiantes con las siguientes columnas:

| Nombre | Edad | Carrera | Promedio |
|---------|------|----------|-----------|
| Ana | 20 | Informática | 9.4 |
| Luis | 21 | Civil | 8.7 |
| Sofía | 19 | Química | 9.8 |

Si seleccionamos únicamente las columnas **Nombre** y **Promedio**, obtendremos un nuevo **DataFrame** que contiene solo esa información, manteniendo todas las filas del conjunto de datos original.

> 🚀 La selección de múltiples columnas es una técnica fundamental para trabajar únicamente con la información necesaria, facilitando el análisis, la limpieza y la manipulación de datos en Pandas.

## 📁 Cargando el conjunto de datos del Titanic
Usaremos los datos del Titanic 🛳️ para aprender a seleccionar más de una columna con pandas.

```python
import pandas as pd
```

📌 Cargamos el archivo CSV y lo guardamos en un DataFrame llamado `titanic`.
Esto nos permitirá analizar la información de los pasajeros.

```python
titanic = pd.read_csv("data/titanic_v5.csv")
titanic
```

## 👀 Visualizando múltiples columnas: `nombre` y `edad`
Seleccionamos solo estas dos columnas y después mostramos las últimas 10 filas del resultado.

```python
titanic[["nombre", "edad"]]
```

```python
titanic[["nombre", "edad"]].tail(10)
```

## 📊 Estadísticas de columnas numéricas: `sobrevivio` y `edad`
Con `.describe()` obtenemos información como la media, desviación estándar, valores mínimo y máximo, etc.

```python
titanic[["sobrevivio", "edad"]].describe()
```

## 🔝 Seleccionar los valores más grandes mezclando columnas
Con `nlargest()` mostramos los 10 registros con valores más altos combinando `clase_pasajero` y `edad`.
Ideal para encontrar adultos mayores en clases altas, por ejemplo.

```python
# Ordenamos por clase de pasajero y edad de forma descendente
titanic.nlargest(10, ["clase_pasajero", "edad"])
```

### 🛠️ Selección usando una lista de columnas
Podemos crear una lista con los nombres de las columnas que nos interesan 📋 y luego usarla para extraerlas del DataFrame.

```python
columnas = ["nombre", "sexo", "edad"]
titanic[columnas]
```
