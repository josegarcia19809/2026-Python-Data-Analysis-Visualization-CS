# Títulos de Netflix
## Búsqueda de datos faltantes y tipos de datos

En este análisis veremos cómo detectar valores faltantes (`NaN`) en un DataFrame y cómo interpretar los tipos de datos de cada columna.

```python
import pandas as pd
```

Leemos el archivo que contiene información sobre títulos disponibles en Netflix:

```python
netflix = pd.read_csv("data/titulos_netflix.csv")
netflix
```

Quizás a simple vista no podamos notar si hay datos faltantes. En Pandas, los datos ausentes se representan como `NaN`, que significa *Not a Number*.

Para revisar si hay columnas con valores ausentes y conocer el tipo de cada una, usamos el método `info()`:

```python
netflix.info()
```

### Interpretación del resultado
- Se observa que el DataFrame tiene 8807 registros.
- Algunas columnas como `director`, `clasificación` y `duración` tienen **menos de 8807 valores no nulos**, lo que indica que tienen datos faltantes (`NaN`).

## ¿Qué tipo de dato tiene cada columna?
Además de los valores no nulos, `info()` también muestra el **tipo de dato** de cada columna.

- Por ejemplo, la columna `año_lanzamiento` es de tipo `int64`, lo que indica que contiene números enteros.
- Las columnas con tipo `object` suelen contener cadenas de texto (strings).

```python
netflix.info()
```

> ✅ **Nota**: El tipo `object` en Pandas normalmente representa datos de texto.
