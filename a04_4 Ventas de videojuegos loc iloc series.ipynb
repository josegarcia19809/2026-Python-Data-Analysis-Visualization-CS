# 🧪 Pandas `loc` e `iloc` con Series

Las **Series** son la estructura de datos unidimensional de **Pandas**, y al igual que los `DataFrame`, permiten acceder a sus elementos de diferentes maneras. Para ello, Pandas proporciona los métodos **`loc`** e **`iloc`**, que facilitan la selección de datos de forma precisa y eficiente.

La principal diferencia entre ambos es que **`loc` accede a los datos mediante las etiquetas del índice**, mientras que **`iloc` utiliza la posición numérica** de cada elemento. Conocer esta diferencia es fundamental para evitar errores y aprovechar al máximo las capacidades de Pandas.

## 🎯 ¿Qué aprenderás?

En esta sección aprenderás a:

- 📌 Comprender la diferencia entre `loc` e `iloc` en una **Series**.
- 📌 Acceder a un elemento utilizando su etiqueta de índice con `loc`.
- 📌 Seleccionar elementos mediante su posición con `iloc`.
- 📌 Obtener rangos de datos utilizando ambos métodos.
- 📌 Utilizar índices personalizados para facilitar el acceso a la información.

## 💡 Ejemplo de uso

Imagina una **Series** que almacena las ventas de diferentes videojuegos:

| Índice | Ventas (millones) |
|--------|--------------------|
| Minecraft | 300 |
| GTA V | 215 |
| Mario Kart 8 | 75 |
| FIFA 24 | 25 |

Con **`loc`** puedes acceder al valor asociado a la etiqueta **"Minecraft"**, mientras que con **`iloc`** puedes obtener el primer elemento indicando simplemente su posición (`0`). Ambos métodos permiten seleccionar uno o varios elementos según las necesidades del análisis.

> 🚀 Dominar **`loc`** e **`iloc`** en **Series** te ayudará a acceder a los datos de forma rápida, escribir código más claro y aprovechar al máximo las herramientas de análisis que ofrece Pandas.

## 🔑 ¿Qué son `loc` e `iloc`?
- `loc[]`: accede por **etiqueta** o índice explícito.
- `iloc[]`: accede por **posición** (como una lista en Python).
¡Vamos a ver ejemplos usando un conjunto de datos de ventas de videojuegos! 🎮

📥 **Importamos la librería pandas** para poder trabajar con estructuras de datos tipo DataFrame o Series.

```python
import pandas as pd
```

📄 **Leemos el archivo CSV** que contiene las ventas de videojuegos y mostramos el DataFrame completo.

```python
ventas = pd.read_csv("data/ventas_videojuegos.csv")
ventas
```

🎯 **Accedemos a la Serie `Titulo`**, que contiene los nombres de los videojuegos.

```python
ventas["Titulo"]
```

🔍 **Usamos `.loc[[7]]` para obtener el título del videojuego con índice 7.**
Este método conserva el tipo `Series` incluso si seleccionamos un solo elemento (por eso usamos doble corchete).

```python
ventas["Titulo"].loc[[7]]
```

📚 **Seleccionamos múltiples títulos** con `.loc[[7, 8, 9]]`.
Esto permite obtener varios valores con base en sus índices.

```python
ventas["Titulo"].loc[[7, 8, 9]]
```

📏 **Seleccionamos un rango de títulos** usando `.loc[2:5]`.
Recuerda que con `loc`, el **último valor sí está incluido** en el resultado.

```python
ventas["Titulo"].loc[2:5]
```

📊 **Contamos cuántos juegos hay por género** con `value_counts()`.
Esto nos da una idea de los géneros más comunes en el conjunto de datos.

```python
ventas["Genero"].value_counts()
```

🥇 **Mostramos los 2 géneros más populares** usando `.iloc[0:2]` para acceder por posición.
`iloc` nos da los elementos por posición dentro del resultado del `value_counts()`.

```python
ventas["Genero"].value_counts().iloc[0:2]
```
