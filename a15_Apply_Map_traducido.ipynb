📄 **Cargamos el archivo CSV del Titanic** y reemplazamos valores '?' por `None`, luego convertimos columnas a tipo `float`.

## The Series apply( ) method

### 📅 Convertir edades de años a días

```python
def years_to_days(yrs):
    return yrs * 365

titanic["age"] * 365

Esta expresión multiplica cada valor de la columna age por 365, obteniendo una estimación de la edad en días.

A diferencia de apply(), esta operación aprovecha las capacidades vectorizadas de Pandas, por lo que suele ser más rápida y sencilla cuando se realizan cálculos matemáticos sobre una columna completa. 🚢📊

### 👥 Clasificar pasajeros por grupo de edad
Se define una función llamada get_age_group() que clasifica una edad en diferentes categorías:

👶 infant: menor de 2 años.
🧒 child: de 2 a 11 años.
🧑 teen: de 12 a 17 años.
👨 adult: de 18 a 49 años.
👴 senior: 50 años o más.

La función apply() se utiliza para aplicar esta clasificación a cada valor de la columna age, creando una nueva columna llamada age_group con el grupo de edad correspondiente para cada pasajero. 🚢

````python
def get_age_group(age):
    if age < 2:
        return "infant"
    elif age < 12:
        return "child"
    elif age < 18:
        return "teen"
    elif age < 50:
        return "adult"
    else:
        return "senior"
````

````python
titanic["age_group"] = titanic["age"].apply(get_age_group)
````

### 📊 Contar pasajeros por grupo de edad

El método `value_counts()` cuenta cuántas veces aparece cada categoría dentro de la columna `age_group`.

El resultado muestra el número de pasajeros que pertenecen a cada grupo de edad (`infant`, `child`, `teen`, `adult` y `senior`), ordenados desde el grupo más frecuente hasta el menos frecuente.

Esta operación es útil para analizar la distribución de edades de los pasajeros y conocer qué grupos de edad estaban más representados en el conjunto de datos del Titanic. 🚢👥

````python
titanic.age_group.value_counts()
````

### 📊 Calcular la tasa de supervivencia por grupo de edad

Esta operación agrupa a los pasajeros según su grupo de edad (`age_group`) y calcula el promedio de la columna `survived` para cada grupo.

Como la columna `survived` suele contener valores de **1** (sobrevivió) y **0** (no sobrevivió), el promedio representa la proporción o porcentaje de pasajeros que sobrevivieron dentro de cada grupo de edad.

Este análisis permite comparar las tasas de supervivencia entre infantes, niños, adolescentes, adultos y personas mayores, ayudando a identificar posibles diferencias en las probabilidades de supervivencia según la edad. 🚢📈

````python
titanic.groupby("age_group").survived.mean()
````

### 📊 Calcular la tasa de supervivencia por grupo de edad y sexo

Esta operación agrupa a los pasajeros utilizando dos criterios: el grupo de edad (`age_group`) y el sexo (`sex`). Posteriormente, calcula el promedio de la columna `survived` para cada combinación de categorías.

Dado que la columna `survived` contiene valores de **1** para los pasajeros que sobrevivieron y **0** para los que no sobrevivieron, el promedio obtenido representa la tasa de supervivencia de cada grupo.

Este análisis permite comparar las probabilidades de supervivencia entre hombres y mujeres dentro de cada grupo de edad, facilitando la identificación de patrones relacionados con la edad y el sexo de los pasajeros. 🚢👥📈

````python
titanic.groupby(["age_group", "sex"]).survived.mean()
````

# Apply con Lambdas

### 💰 Multiplicar la tarifa por 24

Esta operación multiplica cada valor de la columna `fare` por **24**, generando una nueva serie con los resultados.

Pandas realiza esta operación sobre todos los registros de forma vectorizada, aplicando la multiplicación a cada tarifa individual sin necesidad de utilizar ciclos o funciones adicionales.

El resultado puede utilizarse para realizar conversiones, estimaciones o cálculos derivados a partir de la tarifa original de cada pasajero. 🚢📊

````python
titanic["fare"] * 24
````

🧪 **Usamos `apply()` con `lambda`** para aplicar una función personalizada a una columna.

### 💰 Formatear y transformar las tarifas

Esta operación aplica una función anónima (*lambda*) a cada valor de la columna `fare`.

Para cada tarifa, el valor se multiplica por **24** y posteriormente se convierte en una cadena de texto con el símbolo **$** al inicio.

El resultado es una nueva serie con las tarifas transformadas y formateadas como valores monetarios, lo que puede facilitar su visualización en reportes o tablas. 🚢💵

````python
titanic["fare"].apply(lambda x: f"${x*24}")
````

### 🔧 Definir una función para convertir y formatear valores monetarios

Esta función recibe dos parámetros: un número (`num`) y un multiplicador (`multiplier`).

El valor numérico se multiplica por el factor indicado y el resultado se devuelve como una cadena de texto con el símbolo **$** al inicio, simulando un formato monetario.

Esta función puede reutilizarse para realizar diferentes conversiones simplemente cambiando el valor del multiplicador, evitando repetir la misma lógica en varias partes del código. 💰📊

````python
def convert_currency(num, multiplier):
    return f"${num*multiplier}"
````

### 💰 Aplicar una función personalizada con argumentos adicionales

Esta operación aplica la función `convert_currency()` a cada valor de la columna `fare`.

Además del valor de la tarifa, se envía un argumento adicional mediante `args`, indicando que cada tarifa debe multiplicarse por **24** antes de ser formateada como texto monetario.

El resultado es una nueva serie donde cada tarifa aparece convertida y precedida por el símbolo **$**. Este enfoque es útil cuando una función requiere más de un parámetro y se desea reutilizar con distintos valores de configuración. 🚢📊💵

````python
titanic["fare"].apply(convert_currency, args=(24,))
````

````python
def get_range(s):
    return s.max() - s.min()
````

### 📊 Calcular el rango de cada columna

Esta operación aplica la función `get_range()` a cada columna del DataFrame.

La función obtiene el valor máximo y el valor mínimo de una columna, y luego calcula la diferencia entre ambos. Esta diferencia se conoce como **rango** y representa la amplitud de los datos.

El resultado es una serie donde cada valor indica cuánto varían los datos dentro de una columna, lo que permite comparar rápidamente la dispersión de las diferentes variables del conjunto de datos. 📈

````python
df.apply(get_range)
````

### 📊 Calcular el rango de cada columna

Esta operación aplica la función `get_range()` a lo largo de las columnas del DataFrame (`axis=0`).

Para cada columna, la función calcula la diferencia entre el valor máximo y el valor mínimo, obteniendo así el rango de los datos.

El resultado es una serie que muestra el rango de cada columna, permitiendo conocer la amplitud o variabilidad de los valores en cada variable del conjunto de datos. 📈

````python
df.apply(get_range, axis=0)
````

## Apply() w/ DataFrames: Rows

### 📊 Calcular el rango de cada fila

Esta operación aplica la función `get_range()` a lo largo de las filas del DataFrame (`axis=1`).

Para cada fila, se identifica el valor máximo y el valor mínimo entre sus columnas y se calcula la diferencia entre ambos. El resultado representa el rango de valores dentro de cada registro.

La salida es una serie donde cada valor corresponde al rango de una fila, lo que permite analizar la variación de los datos en cada observación del conjunto de datos. 📈

````python
df.apply(get_range, axis=1)
````

````python
def get_fam_size(s):
    fam_size = s.sibsp + s.parch
    if fam_size == 0:
        return "solo"
    elif fam_size < 5:
        return "average"
    else:
        return "large"
````

### 👨‍👩‍👧‍👦 Clasificar pasajeros según el tamaño de su familia

Esta operación aplica la función `get_fam_size()` a cada fila del DataFrame (`axis=1`).

La función calcula el tamaño de la familia sumando los valores de las columnas `sibsp` (hermanos o cónyuges a bordo) y `parch` (padres o hijos a bordo). Con base en este total, asigna una categoría:

- 👤 `solo`: el pasajero viaja sin familiares.
- 👨‍👩‍👧 `average`: el pasajero viaja con una familia de tamaño pequeño o mediano.
- 👨‍👩‍👧‍👦‍👦 `large`: el pasajero viaja con una familia numerosa.

El resultado es una serie que indica la categoría de tamaño familiar correspondiente a cada pasajero del Titanic. 🚢

````python
titanic.apply(get_fam_size, axis=1)
````

### 👨‍👩‍👧‍👦 Contar pasajeros según el tamaño de su familia

Primero se crea una nueva columna llamada `fam_size`, que almacena la categoría de tamaño familiar asignada a cada pasajero.

Posteriormente, se utiliza `value_counts()` para contar cuántos pasajeros pertenecen a cada categoría:

- 👤 `solo`: pasajeros que viajaban sin familiares.
- 👨‍👩‍👧 `average`: pasajeros que viajaban con una familia pequeña o mediana.
- 👨‍👩‍👧‍👦‍👦 `large`: pasajeros que viajaban con una familia numerosa.

El resultado muestra la cantidad de pasajeros en cada grupo, ordenados desde la categoría más frecuente hasta la menos frecuente. Este análisis permite comprender mejor cómo estaba distribuido el tamaño de las familias a bordo del Titanic. 🚢📊

````python
titanic["fam_size"] = titanic.apply(get_fam_size, axis=1)
````

````python
titanic["fam_size"].value_counts()
````

### 📊 Calcular la tasa de supervivencia según el tamaño de la familia

Esta operación agrupa a los pasajeros de acuerdo con la categoría de tamaño familiar almacenada en `fam_size` y calcula el promedio de la columna `survived` para cada grupo.

Como la columna `survived` contiene valores de **1** para los pasajeros que sobrevivieron y **0** para los que no sobrevivieron, el promedio representa la tasa de supervivencia de cada categoría familiar.

El resultado permite comparar las probabilidades de supervivencia entre pasajeros que viajaban solos, con familias pequeñas o medianas, y con familias numerosas, ayudando a identificar si el tamaño de la familia pudo influir en las posibilidades de sobrevivir al desastre del Titanic. 🚢👨‍👩‍👧‍👦📈

````python
titanic.groupby("fam_size").survived.mean()
````

### 📊 Calcular la tasa de supervivencia según el tamaño de la familia y el sexo

Esta operación agrupa a los pasajeros utilizando dos criterios: la categoría de tamaño familiar (`fam_size`) y el sexo (`sex`). Luego, calcula el promedio de la columna `survived` para cada combinación de grupos.

Debido a que la columna `survived` contiene valores de **1** para los pasajeros que sobrevivieron y **0** para los que no sobrevivieron, el promedio obtenido representa la tasa de supervivencia de cada grupo.

Este análisis permite comparar las probabilidades de supervivencia entre hombres y mujeres que viajaban solos, con familias pequeñas o medianas, o con familias numerosas, facilitando la identificación de patrones relacionados con la composición familiar y el sexo de los pasajeros. 🚢👨‍👩‍👧‍👦👥📈

````python
titanic.groupby(["fam_size", "sex"]).survived.mean()
````

### 🚢 Convertir códigos de clase a etiquetas descriptivas

Esta operación utiliza `map()` para reemplazar los valores numéricos de la columna `pclass` por etiquetas más fáciles de interpretar.

Cada clase se transforma de la siguiente manera:

- 🥇 `1` → `1st`
- 🥈 `2` → `2nd`
- 🥉 `3` → `3rd`

El resultado es una nueva serie con nombres descriptivos en lugar de códigos numéricos, lo que facilita la lectura y el análisis de los datos. 📊

````python
titanic["pclass"].map({1:"1st", 2:"2nd", 3:"3rd"})
````

### 👶 Identificar pasajeros menores de edad

Esta operación utiliza `map()` junto con una función anónima (*lambda*) para evaluar cada valor de la columna `age`.

Para cada edad, se verifica si es menor que **18 años**. El resultado es una serie de valores booleanos:

- ✅ `True`: el pasajero es menor de edad.
- ❌ `False`: el pasajero tiene 18 años o más.

Este tipo de transformación es útil para crear indicadores o filtros basados en una condición específica. 🚢📊

````python
titanic["age"].map(lambda a: a < 18)
````

## The applymap( ) method

Esta operación aplica la función `str.upper()` a cada celda de las columnas seleccionadas.
